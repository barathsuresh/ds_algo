### Bubble Sort
- Bubble Sort is a simple sorting algorithm that repeatedly steps through the list, compares adjacent elements, and swaps them if they are in the wrong order. The pass through the list is repeated until the list is sorted.
- Time Complexity: O(n^2)

#### Psuedo Code
```
function bubbleSort(arr):
    n = length(arr)
    for i from 0 to n-1:
        for j from 0 to n-i-2:
            if arr[j] > arr[j+1]:
                swap arr[j] and arr[j+1]
    return arr
```

#### When to use Bubble Sort?
- Bubble Sort is suitable for small datasets or nearly sorted data due to its simplicity and ease of implementation. However, it is inefficient for large datasets compared to more advanced sorting algorithms.

#### When not to use Bubble Sort?
- Avoid using Bubble Sort for large datasets or when performance is critical, as its O(n^2) time complexity makes it significantly slower than more efficient algorithms like Quick Sort or Merge Sort.

````python
def bubbleSort(arr):
    for i in range(len(arr)):
        for j in range(0, len(arr)-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr
arr = [64, 34, 25, 12, 22, 11, 90, 22]
sorted_arr = bubbleSort(arr)
print("Sorted array is:", sorted_arr)
````

### Selection Sort
- Selection Sort is a simple sorting algorithm that divides the input list into two parts: a sorted and an unsorted region. It repeatedly selects the smallest (or largest) element from the unsorted region and moves it to the end of the sorted region.
- Time Complexity: O(n^2)
#### Psuedo Code
```
function selectionSort(arr):
    n = length(arr)
    for i from 0 to n-1:
        minIndex = i
        for j from i+1 to n-1:
            if arr[j] < arr[minIndex]:
                minIndex = j
        swap arr[i] and arr[minIndex]
    return arr
```
#### When to use Selection Sort?
- Selection Sort is useful for small datasets or when memory usage is a concern, as it performs minimal data movement. However, it is not efficient for large datasets due to its O(n^2) time complexity.
#### When not to use Selection Sort?
- Avoid using Selection Sort for large datasets or when performance is critical, as it is significantly slower than more advanced sorting algorithms like Quick Sort or Merge Sort.

````python
def selectionSort(arr):
    for i in range(len(arr)):
        min_index = i
        for j in range(i+1,len(arr)):
            if arr[j] < arr[min_index]:
                min_index = j
        temp = arr[i]
        arr[i] = arr[min_index]
        arr[min_index] = temp
    return arr

arr = [64, 25, 12, 22, 11]
sorted_arr = selectionSort(arr)
print("Sorted array is:", sorted_arr)
````

### Insertion Sort
- Insertion Sort is a simple sorting algorithm that builds a sorted array one element at a time. It works by taking an element from the unsorted portion and inserting it into its correct position in the sorted portion.
- Time Complexity: O(n^2)
#### Psuedo Code
```
function insertionSort(arr):
    n = length(arr)
    for i from 1 to n-1:
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j = j - 1
        arr[j + 1] = key
    return arr
```
#### When to use Insertion Sort?
- Insertion Sort is efficient for small datasets or nearly sorted data, as it has a low overhead and performs well in these scenarios. It is also stable and easy to implement.
#### When not to use Insertion Sort?
- Avoid using Insertion Sort for large datasets or when performance is critical, as its O(n^2) time complexity makes it significantly slower than more efficient algorithms like Quick Sort or Merge Sort.

````python
def insertionSort(arr):
    for i in range(1,len(arr)):
        key = arr[i]
        j=i-1
        while j>=0 and key<arr[j]:
            arr[j+1] = arr[j]
            j=j-1
        arr[j+1] = key
    return arr

arr = [12, 11, 13, 5, 6]
sorted_arr = insertionSort(arr)
print("Sorted array is:", sorted_arr)
````

### Bucket Sort
- Bucket Sort is a sorting algorithm that distributes the elements of an array into a number of buckets. Each bucket is then sorted individually, either using a different sorting algorithm or by recursively applying the bucket sort algorithm. Finally, the sorted buckets are concatenated to produce the final sorted array.
- Time Complexity: O(n + k) where n is the number of elements and k is the number of buckets.
#### Psuedo Code
```
function bucketSort(arr, numBuckets):
    create an array of empty buckets
    for each element in arr:
        index = floor(numBuckets * element / maxValue)
        add element to buckets[index]
    for each bucket in buckets:
        sort(bucket) // using another sorting algorithm
    return concatenate all buckets
```
#### When to use Bucket Sort?
- Bucket Sort is effective for uniformly distributed data over a known range. It can achieve linear time complexity in such cases, making it suitable for large datasets with a limited range of values.
#### When not to use Bucket Sort?
- Avoid using Bucket Sort for datasets with a wide range of values or non-uniform distributions, as it may lead to uneven bucket sizes and degrade performance.

````python
import math

def bucketSort(arr):
    numberofBuckets = round(math.sqrt(len(arr)))
    minValue = min(arr)
    maxValue = max(arr)
    rangeVal = (maxValue - minValue) / numberofBuckets
 
    buckets = [[] for _ in range(numberofBuckets)]
 
    for j in arr:
        if j == maxValue:
            buckets[-1].append(j)
        else:
            index_b = math.floor((j - minValue) / rangeVal)
            buckets[index_b].append(j)
    
    sorted_array = []
    for i in range(numberofBuckets):
        buckets[i] = insertionSort(buckets[i])
        sorted_array.extend(buckets[i])
    
    return sorted_array

arr = [0.78, -0.17, 0.39, 0.26, 0.72, 0.94, 0.21, 0.55, 0.43, 0.88]
sorted_arr = bucketSort(arr)
print("Sorted array is:", sorted_arr)
````

### Merge Sort
- Merge Sort is a divide-and-conquer sorting algorithm that divides the input array into two halves, recursively sorts each half, and then merges the sorted halves back together.
- Time Complexity: O(n log n)
#### Psuedo Code
```
function mergeSort(arr):
    if length(arr) <= 1:
        return arr
    mid = length(arr) / 2
    left = mergeSort(arr[0:mid])
    right = mergeSort(arr[mid:length(arr)])
    return merge(left, right)
function merge(left, right):
    result = empty array
    while left is not empty and right is not empty:
        if left[0] <= right[0]:
            append left[0] to result
            remove left[0] from left
        else:
            append right[0] to result
            remove right[0] from right
    append remaining elements of left to result
    append remaining elements of right to result
    return result
```
#### When to use Merge Sort?
- Merge Sort is suitable for large datasets and linked lists, as it has a consistent O(n log n) time complexity. It is also stable and works well for external sorting (sorting data that does not fit into memory).
#### When not to use Merge Sort?
- Avoid using Merge Sort for small datasets or when memory usage is a concern, as it requires additional space for the temporary arrays used during the merge process.

````python
def merge(arr,left,mid,right):
    n1 = mid - left + 1 # size of left subarray
    n2 = right - mid    # size of right subarray
    leftArr = [0]*n1
    rightArr = [0]*n2

    for i in range(0,n1):
        leftArr[i] = arr[left + i]
    for j in range(0,n2):
        rightArr[j] = arr[mid + 1 + j]
    
    i=0
    j=0
    k=left
    while i<n1 and j<n2:
        if leftArr[i] <= rightArr[j]:
            arr[k] = leftArr[i]
            i+=1
        else:
            arr[k] = rightArr[j]
            j+=1
        k+=1
    while i<n1:
        arr[k] = leftArr[i]
        i+=1
        k+=1
    while j<n2:
        arr[k] = rightArr[j]
        j+=1
        k+=1

def mergeSort(arr,left,right):
    if left < right:
        mid = (left + right)//2
        mergeSort(arr,left,mid)
        mergeSort(arr,mid+1,right)
        merge(arr,left,mid,right)

arr = [38, 27, 43, 3, 9, 82, 10]
mergeSort(arr,0,len(arr)-1)
print("Sorted array is:", arr)
````

### Quick Sort
- Quick Sort is a divide-and-conquer sorting algorithm that selects a 'pivot' element from the array and partitions the other elements into two sub-arrays according to whether they are less than or greater than the pivot. The sub-arrays are then sorted recursively.
- Time Complexity: O(n log n) on average, O(n^2) in the worst case.
#### Psuedo Code
```
function quickSort(arr):
    if length(arr) <= 1:
        return arr
    pivot = arr[length(arr) / 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quickSort(left) + middle + quickSort(right)
```
#### When to use Quick Sort?
- Quick Sort is efficient for large datasets and is often faster in practice than other O(n log n) algorithms like Merge Sort and Heap Sort. It is suitable for in-memory sorting and has good cache performance.
#### When not to use Quick Sort?
- Avoid using Quick Sort for small datasets or when stability is required, as it is not a stable sort. Additionally, be cautious with datasets that may lead to the worst-case O(n^2) performance, such as already sorted arrays.

````python
def swap(arr,i,j):
    arr[i], arr[j] = arr[j], arr[i]

def pivot(arr,pivot_index,end_index):
    swap_index = pivot_index
    for i in range(pivot_index+1,end_index+1):
        if arr[i] < arr[pivot_index]:
            swap_index += 1
            swap(arr,swap_index,i)
    swap(arr,pivot_index,swap_index)
    return swap_index

def quickSort(arr,left,right):
    if left < right:
        pivot_index = pivot(arr,left,right)
        quickSort(arr,left,pivot_index-1)
        quickSort(arr,pivot_index+1,right)

arr = [10, 7, 8, 9, 1, 5]
quickSort(arr,0,len(arr)-1)
print("Sorted array is:", arr)
````

### Heap Sort
- Heap Sort is a comparison-based sorting algorithm that uses a binary heap data structure to sort elements. It first builds a max heap from the input data, then repeatedly extracts the maximum element from the heap and rebuilds the heap until all elements are sorted.
- Time Complexity: O(n log n)
#### Psuedo Code
```
function heapSort(arr):
    buildMaxHeap(arr)
    for i from length(arr)-1 down to 1:
        swap arr[0] and arr[i]
        heapify(arr, 0, i)
    return arr
function buildMaxHeap(arr):
    n = length(arr)
    for i from floor(n/2)-1 down to 0:
        heapify(arr, i, n)
function heapify(arr, i, n):
    largest = i
    left = 2*i + 1
    right = 2*i + 2
    if left < n and arr[left] > arr[largest]:
        largest = left
    if right < n and arr[right] > arr[largest]:
        largest = right
    if largest != i:
        swap arr[i] and arr[largest]
        heapify(arr, largest, n)
```
#### When to use Heap Sort?
- Heap Sort is suitable for large datasets and when constant space usage is required, as it sorts
    in place. It has a consistent O(n log n) time complexity and does not require additional memory for recursive calls.
#### When not to use Heap Sort?
- Avoid using Heap Sort when stability is required, as it is not a stable sort. Additionally, it may have slower performance compared to Quick Sort for average cases due to its memory access patterns.

````python
def heapify(arr,n,i):
    largest = i
    left = 2*i + 1
    right = 2*i + 2

    if left < n and arr[left] > arr[largest]:
        largest = left
    if right < n and arr[right] > arr[largest]:
        largest = right
    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]
        heapify(arr,n,largest)
def buildMaxHeap(arr):
    n = len(arr)
    for i in range(n//2 - 1, -1, -1):
        heapify(arr,n,i)
def heapSort(arr):
    n = len(arr)
    buildMaxHeap(arr)
    for i in range(n-1,0,-1):
        arr[i], arr[0] = arr[0], arr[i]
        heapify(arr,i,0)
arr = [12, 11, 13, 5, 6, 7]
heapSort(arr)
print("Sorted array is:", arr)
````

# Searching Algorithms
## Types of Searching Algorithms
- Linear Search
- Binary Search

````python
def linearSearch(arr,target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1
arr = [2, 3, 4, 10, 40]
target = 10
result = linearSearch(arr,target)
if result != -1:
    print("Element is present at index", str(result))
else:
    print("Element is not present in array")
````

````python
def binarySearch(arr,target):
    left = 0
    right = len(arr) - 1
    while left <= right:
        mid = left + (right - left) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1
arr = [2, 3, 4, 10, 40]
target = 10
result = binarySearch(arr,target)
if result != -1:
    print("Element is present at index", str(result))
else:
    print("Element is not present in array")
````

# Topological Sorting
- Topological Sorting is a linear ordering of vertices in a directed acyclic graph (DAG) such that for every directed edge (u, v) from vertex u to vertex v, u comes before v in the ordering. It is commonly used in scenarios like task scheduling, where certain tasks must be completed before others can begin.
## When to use Topological Sorting?
- Topological Sorting is useful for scheduling tasks with dependencies, resolving symbol dependencies in linkers, and organizing data with hierarchical relationships. It is applicable in scenarios where a linear order of execution is required based on precedence constraints.
## When not to use Topological Sorting?
- Topological Sorting is not applicable for graphs that contain cycles, as it requires a directed acyclic graph (DAG) to produce a valid linear ordering. If the graph has cycles, topological sorting cannot be performed.

````python
from collections import deque
class Graph:
    def __init__(self,gDict = None):
        if gDict is None:
            gDict = {}
        self.graph = gDict
    
    def addEdge(self,vertex1,vertex2):
        if vertex1 not in self.graph.keys():
            self.graph[vertex1] = []
        if vertex2 not in self.graph.keys():
            self.graph[vertex2] = []
        self.graph[vertex1].append(vertex2)
        self.graph[vertex2].append(vertex1)
    
    def addVertex(self,vertex):
        if vertex not in self.graph.keys():
            self.graph[vertex] = []
    
    def removeEdge(self,vertex1,vertex2):
        if vertex1 in self.graph.keys() and vertex2 in self.graph.keys():
            if vertex2 in self.graph[vertex1]:
                self.graph[vertex1].remove(vertex2)
            if vertex1 in self.graph[vertex2]:
                self.graph[vertex2].remove(vertex1)
    
    def removeVertex(self,vertex):
        if vertex in self.graph.keys():
            del self.graph[vertex]
        for v in self.graph:
            if vertex in self.graph[v]:
                self.graph[v].remove(vertex)

    def printGraph(self):
        for vertex in self.graph:
            print(vertex, ":", self.graph[vertex])
    
    def bfs(self,vertex):
        visited = set()
        queue = deque([vertex])
        visited.add(vertex)
        while queue:
            current = queue.popleft()
            print(current, end=" ")
            for neighbor in self.graph[current]:
                if neighbor not in visited:
                    visited.add(neighbor)
                    queue.append(neighbor)
    
    def dfs_recur(self,vertex,visited=None):
        if visited is None:
            visited = set()
        visited.add(vertex)
        print(vertex, end=" ")
        for neighbor in self.graph[vertex]:
            if neighbor not in visited:
                self.dfs_recur(neighbor,visited)
    
    def dfs(self,vertex):
        visited = set()
        stack = [vertex]
        while stack:
            current = stack.pop()
            if current not in visited:
                print(current, end=" ")
                visited.add(current)
                for neighbor in self.graph[current]:
                    if neighbor not in visited:
                        stack.append(neighbor)
    
    def topologicalSort(self):
        visited = set()
        stack = []
        for vertex in self.graph:
            if vertex not in visited:
                self._topologicalSortUtil(vertex,visited,stack)
        while stack:
            print(stack.pop(), end=" ")

    def _topologicalSortUtil(self,vertex,visited,stack):
        visited.add(vertex)
        for neighbor in self.graph[vertex]:
            if neighbor not in visited:
                self._topologicalSortUtil(neighbor,visited,stack)
        stack.append(vertex)
    
    def sssppBfs(self,source,destination):
        queue = []
        queue.append([source])
        while queue:
            path = queue.pop(0)
            node = path[-1]
            if node == destination:
                return path
            for adjacent in self.graph.get(node, []):
                new_path = list(path)
                new_path.append(adjacent)
                queue.append(new_path)

    
# Create a complex graph with letter vertices
g = Graph()

# Main connected component
g.addEdge("A", "B")
g.addEdge("A", "C") 
g.addEdge("A", "D")
g.addEdge("B", "E")
g.addEdge("B", "F")
g.addEdge("C", "G")
g.addEdge("D", "H")
g.addEdge("E", "I")
g.addEdge("F", "J")
g.addEdge("G", "K")
g.addEdge("H", "L")
g.addEdge("I", "J")  # Creates cycle
g.addEdge("K", "L")  # Creates cycle

# Secondary connected component
g.addEdge("M", "N")
g.addEdge("N", "O")
g.addEdge("O", "P")
g.addEdge("P", "Q")
g.addEdge("Q", "M")  # Creates cycle

# Third component (smaller)
g.addEdge("R", "S")
g.addEdge("S", "T")
g.addEdge("T", "R")  # Triangle

# Isolated vertices
g.addVertex("U")
g.addVertex("V")
g.addVertex("W")

# Print and test
g.printGraph()
print("\nBFS from A:")
g.bfs("A")
print("\nBFS from M:")
g.bfs("M")
print("\nBFS from R:")
g.bfs("R")

# Test DFS implementations
print("\n\nDFS RECURSIVE from A:")
g.dfs_recur("A")
print("\n\nDFS ITERATIVE from A:")
g.dfs("A")

print("\n\nDFS RECURSIVE from M:")
g.dfs_recur("M")
print("\n\nDFS ITERATIVE from M:")
g.dfs("M")

print("\n\nDFS RECURSIVE from R:")
g.dfs_recur("R")
print("\n\nDFS ITERATIVE from R:")
g.dfs("R")

print("\n\nTopological Sort of the graph:")
g.topologicalSort()

print("\n\nShortest path from A to J:")
path = g.sssppBfs("A","J")
print(" -> ".join(path) if path else "No path found")
````

# Dijkstra's Algorithm
- Dijkstra's Algorithm is a popular algorithm used to find the shortest path from a starting vertex to all other vertices in a weighted graph with non-negative edge weights. It works by iteratively selecting the vertex with the smallest known distance, updating the distances of its neighbors, and repeating this process until all vertices have been processed.
## Psuedo Code
```
function dijkstra(graph, start):
    create a priority queue Q
    create a map dist to store the shortest distance to each vertex
    for each vertex v in graph:
        dist[v] = infinity
        add v to Q
    dist[start] = 0
    while Q is not empty:
        u = vertex in Q with smallest dist[u]
        remove u from Q
        for each neighbor v of u:
            alt = dist[u] + length(u, v)
            if alt < dist[v]:
                dist[v] = alt
    return dist
```
## When to use Dijkstra's Algorithm?
- Dijkstra's Algorithm is suitable for finding the shortest paths in graphs with non-negative edge weights, such as road networks or communication networks. It is efficient for dense graphs and can be used in applications like GPS navigation and network routing.
## When not to use Dijkstra's Algorithm?
- Avoid using Dijkstra's Algorithm for graphs with negative edge weights, as it may produce incorrect results. In such cases, consider using the Bellman-Ford algorithm, which can handle negative weights and detect negative cycles.

````python
import heapq
class Edge:
    def __init__(self,weight,start_vertex,target_vertex):
        self.weight = weight
        self.start_vertex = start_vertex
        self.target_vertex = target_vertex

class Node:
    def __init__(self,vertex):
        self.vertex = vertex
        self.visited = False
        # previous node that we come from in path
        self.predecessor = None
        self.neighbors = []
        self.min_distance = float('inf')
    # we use this for min heap implementation  
    def __lt__(self,other):
        return self.min_distance < other.min_distance

    def add_edge(self,weight,destination_vertex):
        ed = Edge(weight,self,destination_vertex)
        self.neighbors.append(ed)

class Dijkstra:
    def __init__(self):
        self.heap = []

    def calculate(self,start_vertex):
        start_vertex.min_distance = 0
        heapq.heappush(self.heap,start_vertex)
        while self.heap:
            current_vertex = heapq.heappop(self.heap)
            if current_vertex.visited:
                continue
            for edge in current_vertex.neighbors:
                u = edge.start_vertex
                v = edge.target_vertex
                new_distance = u.min_distance + edge.weight
                if new_distance < v.min_distance:
                    v.predecessor = u
                    v.min_distance = new_distance
                    heapq.heappush(self.heap,v)
            current_vertex.visited = True
    
    def get_shortest_path(self,target_vertex):
        path = []
        current_vertex = target_vertex
        while current_vertex is not None:
            path.append(current_vertex.vertex)
            current_vertex = current_vertex.predecessor
        path.reverse()
        return path

# Example usage:
a = Node("A")
b = Node("B")
c = Node("C")
d = Node("D")
e = Node("E")
a.add_edge(1,b)
a.add_edge(4,c)
b.add_edge(2,c)
b.add_edge(5,d)
c.add_edge(1,d)
d.add_edge(3,e)
dijkstra = Dijkstra()
dijkstra.calculate(a)
path = dijkstra.get_shortest_path(e)
print("Shortest path from A to E:", " -> ".join(path))

    
````

# Bellman-Ford Algorithm
- Bellman-Ford Algorithm is a graph algorithm used to find the shortest paths from a single source vertex to all other vertices in a weighted graph, even when the graph contains edges with negative weights. It works by iteratively relaxing the edges of the graph, updating the shortest path estimates until no further improvements can be made.
## Psuedo Code
```
function bellmanFord(graph, start):
    create a map dist to store the shortest distance to each vertex
    for each vertex v in graph:
        dist[v] = infinity
    dist[start] = 0
    for i from 1 to number of vertices - 1:
        for each edge (u, v) in graph:
            if dist[u] + weight(u, v) < dist[v]:
                dist[v] = dist[u] + weight(u, v)
    for each edge (u, v) in graph:
        if dist[u] + weight(u, v) < dist[v]:
            error "Graph contains a negative-weight cycle"
    return dist
```
## When to use Bellman-Ford Algorithm?
- Bellman-Ford Algorithm is suitable for graphs with negative edge weights, as it can correctly find the shortest paths and detect negative cycles. It is useful in applications like currency exchange rate calculations and network routing with varying costs.
## When not to use Bellman-Ford Algorithm?
- Avoid using Bellman-Ford Algorithm for graphs with only non-negative edge weights, as Dijkstra's Algorithm is more efficient in such cases. Bellman-Ford has a higher time complexity of O(VE) compared to Dijkstra's O(E + V log V) with a priority queue.
## Time complexity
- Dijkstra's Algorithm: O(E + V log V) with a priority queue
- Bellman-Ford Algorithm: O(VE)
## Why Bellman for runs v-1 times
- The Bellman-Ford algorithm relaxes all edges in the graph V-1 times (where V is the number of vertices) because, in the worst-case scenario, the longest possible path between any two vertices in a graph can consist of V-1 edges. By performing V-1 iterations of edge relaxation, the algorithm ensures that the shortest paths are correctly computed for all vertices, even in the presence of negative edge weights.

````python
class BellmanFord:
    def __init__(self,vertices):
        self.V = vertices # No. of vertices
        self.graph = []
        self.nodes = []
    
    def add_edge(self,s,d,w):
        self.graph.append((s,d,w))
    
    def add_node(self,vertex):
        self.nodes.append(vertex)
    
    def print_solution(self,dist):
        print("Vertex Distance from Source")
        for key,value in dist.items():
            print(f"{key}\t\t{value}")
    
    def bellman_ford(self,src):
        dist = {i:float("inf") for i in self.nodes}
        dist[src] = 0
        for _ in range(self.V-1):
            for u,v,w in self.graph:
                if  dist[u] != float("inf") and dist[u] + w < dist[v]:
                    dist[v] = dist[u] + w
        
        for u,v,w in self.graph:
            if dist[u] != float("inf") and dist[u] + w < dist[v]:
                print("Graph contains negative weight cycle")
                return
        self.print_solution(dist)

# Example usage:
# Example usage:
bf = BellmanFord(4)
bf.add_node("A")
bf.add_node("B")  # Fixed: Added missing closing parenthesis
bf.add_node("C")
bf.add_node("D")
bf.add_edge("A","B",1)
bf.add_edge("B","C",-2)
bf.add_edge("A","C",4)
bf.add_edge("C","D",2)
bf.add_edge("D","B",-1)
bf.bellman_ford("A") # for negative cycle detection

bf1 = BellmanFord(4)
bf1.add_node("A")
bf1.add_node("B")
bf1.add_node("C")
bf1.add_node("D")
bf1.add_edge("A","B",1)
bf1.add_edge("B","C",2)
bf1.add_edge("A","C",4)
bf1.add_edge("C","D",2)
bf1.bellman_ford("A")
````

# Floyd-Warshall Algorithm
- Floyd-Warshall Algorithm is a dynamic programming algorithm used to find the shortest paths between all pairs of vertices in a weighted graph. It works by iteratively updating a distance matrix that represents the shortest distances between each pair of vertices, considering each vertex as an intermediate point.
## Psuedo Code
```
function floydWarshall(graph):
    n = number of vertices in graph
    create a 2D array dist[n][n]
    for each vertex i in graph:
        for each vertex j in graph:
            if i == j:
                dist[i][j] = 0
            else if edge (i, j) exists:
                dist[i][j] = weight(i, j)
            else:
                dist[i][j] = infinity
    for k from 0 to n-1:
        for i from 0 to n-1:
            for j from 0 to n-1:
                if dist[i][k] + dist[k][j] < dist[i][j]:
                    dist[i][j] = dist[i][k] + dist[k][j]
    return dist
```
## When to use Floyd-Warshall Algorithm?
- Floyd-Warshall Algorithm is suitable for dense graphs and when you need to find shortest paths between all pairs of vertices. It is useful in applications like network routing, social network analysis, and transportation planning.
## When not to use Floyd-Warshall Algorithm?    
- Avoid using Floyd-Warshall Algorithm for sparse graphs or when only single-source shortest paths are needed, as its O(V^3) time complexity can be inefficient compared to other algorithms like Dijkstra's or Bellman-Ford for those scenarios.
## Time Complexity
- Floyd-Warshall Algorithm: O(V^3)
## Floyd-Warshall in a negative weight cycle
- The Floyd-Warshall algorithm can detect negative weight cycles in a graph by checking the diagonal of the distance matrix after completing the algorithm. If any entry dist[i][i] (the distance from a vertex to itself) becomes negative, it indicates the presence of a negative weight cycle in the graph. This is because a negative value on the diagonal implies that there is a path starting and ending at the same vertex with a total negative weight, which is characteristic of a negative weight cycle.

# Disjoint Set Union (DSU)- Disjoint Set Union (DSU), also known as Union-Find, is a data structure
- Disjoint Set Union (DSU), also known as Union-Find, is a data structure that keeps track of a partition of a set into disjoint subsets. It supports two primary operations: "find," which determines the representative (or root) of the subset containing a particular element, and "union," which merges two subsets into a single subset.
## Psuedo Code
```
class DSU:
    function __init__(n):
        parent = array of size n
        rank = array of size n
        for i from 0 to n-1:
            parent[i] = i
            rank[i] = 0
    function find(x):
        if parent[x] != x:
            parent[x] = find(parent[x]) // path compression
        return parent[x]
    function union(x, y):
        rootX = find(x)
        rootY = find(y)
        if rootX != rootY:
            if rank[rootX] > rank[rootY]:
                parent[rootY] = rootX
            else if rank[rootX] < rank[rootY]:
                parent[rootX] = rootY
            else:
                parent[rootY] = rootX
                rank[rootX] += 1
```
## When to use Disjoint Set Union (DSU)?
- Disjoint Set Union (DSU) is useful in scenarios that require efficient management of dynamic connectivity, such as network connectivity, image processing, and Kruskal's algorithm for finding minimum spanning trees. It is particularly effective when frequent union and find operations are needed.
## When not to use Disjoint Set Union (DSU)?
- Avoid using Disjoint Set Union (DSU) when the problem does not involve dynamic connectivity or when the data structure's overhead is unnecessary. For static sets or when only a few union/find operations are required, simpler data structures may be more appropriate.
## Time Complexity
- The time complexity for both "find" and "union" operations in a Disjoint Set Union (DSU) data structure is nearly constant time, specifically O(α(n)), where α is the inverse Ackermann function. This is achieved through optimizations such as path compression and union by rank, which significantly improve the efficiency of these operations over a series of union and find calls.

````python
class DisjointSet:
    def __init__(self,vertices):
        self.vertices = vertices
        self.parent = {}
        for v in vertices:
            self.parent[v] = v
        self.rank = dict.fromkeys(vertices,0)

    def find(self,item):
        if self.parent[item] == item:
            return item
        else:
            return self.find(self.parent[item])
    
    def union(self,set1,set2):
        xroot = self.find(set1)
        yroot = self.find(set2)
        if self.rank[xroot] < self.rank[yroot]:
            self.parent[xroot] = yroot
        elif self.rank[xroot] > self.rank[yroot]:
            self.parent[yroot] = xroot
        else:
            self.parent[yroot] = xroot
            self.rank[xroot] += 1

# Example usage:
vertices = ['A','B','C','D','E']
ds = DisjointSet(vertices)
ds.union('B','A')
ds.union('C','D')
print("Find A:", ds.find('A'))
print("Find B:", ds.find('B'))
print("Find C:", ds.find('C'))
print("Find D:", ds.find('D'))
print("Find E:", ds.find('E'))
print("Parents:", ds.parent)


    
````

# Minimum Spanning Tree (MST)
- A Minimum Spanning Tree (MST) is a subset of the edges of a connected, undirected graph that connects all the vertices together without any cycles and with the minimum possible total edge weight. In other words, it is a tree that spans all the vertices in the graph while minimizing the sum of the weights of the edges included in the tree.
## Algorithms to find MST
- Kruskal's Algorithm
- Prim's Algorithm
## When to use Minimum Spanning Tree (MST)?
- Minimum Spanning Tree (MST) is useful in scenarios such as network design (e.g., designing least-cost communication or transportation networks), clustering analysis, and circuit design. It helps in minimizing the total cost while ensuring connectivity among all nodes.
## When not to use Minimum Spanning Tree (MST)?
- Avoid using Minimum Spanning Tree (MST) when the graph is directed, as MST algorithms are designed for undirected graphs. Additionally, if the goal is to find the shortest path between specific nodes rather than connecting all nodes, other algorithms like Dijkstra's or Bellman-Ford may be more appropriate.
## Time Complexity
- Kruskal's Algorithm: O(E log E) or O(E log V), where E is the number of edges and V is the number of vertices.
- Prim's Algorithm: O(E + V log V) with a priority queue, where E is the number of edges and V is the number of vertices.

## Kruskal's Algorithm
- Kruskal's Algorithm is a greedy algorithm used to find the Minimum Spanning Tree (MST) of a connected, undirected graph. It works by sorting all the edges of the graph in non-decreasing order of their weights and then adding edges to the MST one by one, ensuring that no cycles are formed, until all vertices are connected.
### Psuedo Code
```
function kruskal(graph):
    create a list of edges from graph
    sort edges in non-decreasing order of weight
    create a DSU to manage connected components
    mst = empty list
    for each edge (u, v) in sorted edges:
        if find(u) != find(v): // no cycle
            union(u, v)
            add edge (u, v) to mst
    return mst
```
### When to use Kruskal's Algorithm?
- Kruskal's Algorithm is suitable for sparse graphs with a large number of vertices and relatively few edges. It is efficient when the edges can be easily sorted and when the graph is not densely connected.
### When not to use Kruskal's Algorithm?
- Avoid using Kruskal's Algorithm for dense graphs, as Prim's Algorithm may be more efficient in such cases. Additionally, if the graph is not connected, Kruskal's Algorithm will only produce a minimum spanning forest rather than a single MST.

````python
class KruskalMST:
    def __init__(self,vertices):
        self.NumVertices = vertices
        self.graph = []
        self.nodes = []
        self.MST = []

    def add_edge(self,u,v,w):
        self.graph.append((u,v,w))
    
    def add_node(self,value):
        self.nodes.append(value)
    
    def print_MST(self):
        print("Edges in the Minimum Spanning Tree:")
        for w,u,v in self.MST:
            print(f"{u} -- {v} == {w}")
    
    def kruskal(self):
        ds = DisjointSet(self.nodes)
        self.graph = sorted(self.graph,key=lambda item: item[2])
        i,e = 0,0
        while e < self.NumVertices - 1:
            u,v,w = self.graph[i]
            i+=1
            x = ds.find(u)
            y= ds.find(v)
            if x!= y:
                e+=1
                self.MST.append((w,u,v))
                ds.union(x,y)
        self.print_MST()
        print("Disjoint Set Parents after Kruskal's Algorithm:", ds.parent)

# Example usage:
kruskal =KruskalMST(4)
kruskal.add_node("A")
kruskal.add_node("B")
kruskal.add_node("C")
kruskal.add_node("D")
kruskal.add_edge("A","B",10)
kruskal.add_edge("A","C",6)
kruskal.add_edge("A","D",5)
kruskal.add_edge("B","D",15)
kruskal.add_edge("C","D",4)
kruskal.kruskal()
````

# Greedy Algorithms
- Greedy Algorithms are a class of algorithms that make a series of choices, each of which looks best at the moment, with the hope of finding a global optimum. They build up a solution piece by piece, always choosing the next piece that offers the most immediate benefit.
## When to use Greedy Algorithms?
- Greedy Algorithms are suitable for optimization problems where local optimal choices lead to a global optimum. They are often used in problems like activity selection, coin change, Huffman coding, and minimum spanning trees.
## When not to use Greedy Algorithms?
- Avoid using Greedy Algorithms for problems where local optimal choices do not guarantee a global optimum, such as the traveling salesman problem or certain knapsack problems. In such cases, dynamic programming or backtracking may be more appropriate.
## Examples of Greedy Algorithms
- Activity Selection Problem
- Coin Change Problem
