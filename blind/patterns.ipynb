## Sliding Window
**Use When**: You are looking for a contiguous sequence of elements in an array or string that satisfies a certain condition.
**The Trick**: Use two pointers to create a "window" that can expand and contract as needed to find the desired sequence. 

```python
def sliding_window(nums):
    l = 0
    current_val = 0  # Sum, Count, Hashmap, etc.
    best_res = 0     # Max length, Min length, etc.
    
    for r in range(len(nums)):
        # 1. EXPAND: Add right element to current state
        current_val += nums[r] 
        
        # 2. SHRINK: While condition is broken (or met, depending on problem)
        while current_val > LIMIT: # Replace with your condition
            current_val -= nums[l] # Remove left element
            l += 1                 # Move left pointer
        
        # 3. RECORD: Update the result
        best_res = max(best_res, r - l + 1)
        
    return best_res
```

## Two Pointers
**Use When**: You need to find pairs in a sorted array that satisfy a certain condition (e.g., sum to a specific value).
**The Trick**: Use two pointers, one starting at the beginning of the array and the other at the end, and move them towards each other based on the condition you're checking.

```python
def two_pointers(nums, target):
    l, r = 0, len(nums) - 1
    
    while l < r:
        current_sum = nums[l] + nums[r]
        
        if current_sum == target:
            return [l, r]      # Found it
        elif current_sum < target:
            l += 1             # Too small? Move Left pointer up to increase sum
        else:
            r -= 1             # Too big? Move Right pointer down to decrease sum
            
    return -1
```

## Fast and Slow Pointers
**Use When**: You need to detect cycles in a linked list or find the middle element of a linked list.
**The Trick**: Use two pointers, one moving twice as fast as the other. If there is a cycle, the fast pointer will eventually meet the slow pointer. If there is no cycle, the fast pointer will reach the end of the list.

```python
def has_cycle(head):
    slow, fast = head, head
    
    # Check fast.next to avoid NullPointer errors
    while fast and fast.next:
        slow = slow.next          # Move 1 step
        fast = fast.next.next     # Move 2 steps
        
        if slow == fast:
            return True           # Collision = Cycle detected
            
    return False
```

## Binary Search
**Use When**: You need to find the position of a target value within a sorted array.
**The Trick**: Repeatedly divide the search interval in half. If the value of the search key is less than the item in the middle of the interval, narrow the interval to the lower half. Otherwise, narrow it to the upper half. Continue until the value is found or the interval is empty.

```python
def binary_search(nums, target):
    l, r = 0, len(nums) - 1
    
    while l <= r:
        mid = (l + r) // 2
        
        if nums[mid] == target:
            return mid          # Found
        elif nums[mid] < target:
            l = mid + 1         # Target is in the right half
        else:
            r = mid - 1         # Target is in the left half
            
    return -1
```

## BFS (Breadth-First Search)
**Use When**: You need to explore all the nodes in a graph or tree level by level.
**The Trick**: Use a queue to keep track of the nodes to be explored next.

```python
from collections import deque

def bfs(root):
    if not root: return []
    
    queue = deque([root])
    result = []
    
    while queue:
        level_size = len(queue)  # Snapshot current size for level-by-level processing
        current_level = []
        
        for _ in range(level_size):
            node = queue.popleft()
            current_level.append(node.val)
            
            # Add neighbors/children
            if node.left: queue.append(node.left)
            if node.right: queue.append(node.right)
            
        result.append(current_level)
    return result
```

## DFS (Depth-First Search)
**Use When**: You need to explore all the nodes in a graph or tree by going as deep as possible before backtracking.
**The Trick**: Use a stack (or recursion) to keep track of the nodes to be explored next.

```python
def solve_backtrack(nums):
    res = []
    
    def backtrack(start_index, current_path):
        # 1. GOAL: Are we done?
        if len(current_path) == len(nums): # or any other base case
            res.append(current_path[:])    # Make a copy!
            return

        # 2. CHOICE: Iterate through options
        for i in range(start_index, len(nums)):
            # (Optional) Pruning: if condition_bad: continue
            
            # CHOOSE
            current_path.append(nums[i])
            
            # EXPLORE (Recurse)
            backtrack(i + 1, current_path)
            
            # UNCHOOSE (Backtrack/Cleanup)
            current_path.pop()
            
    backtrack(0, [])
    return res
```

## Top K Elements
**Use When**: You need to find the k largest or smallest elements in an array.
**The Trick**: Use a min-heap (for largest) or max-heap (for smallest) to efficiently keep track of the top k elements as you iterate through the array.

```python
import heapq

def top_k_elements(nums, k):
    min_heap = []
    
    for n in nums:
        heapq.heappush(min_heap, n)
        
        # If we have more than k elements, pop the smallest
        # This leaves us with the largest K elements in the heap
        if len(min_heap) > k:
            heapq.heappop(min_heap)
            
    return list(min_heap) # Returns the top k largest
```

## Grid Traversal (DFS on Matrix)
**Use When**: You need to explore all the cells in a grid (2D array) that are connected and satisfy a certain condition (e.g., same value).
**The Trick**: Use DFS to explore the grid, marking visited cells to avoid cycles and ensure you only count connected components once.

```python
def grid_dfs(grid):
    ROWS, COLS = len(grid), len(grid[0])
    visited = set() # Or modify grid in-place to '#'
    
    def dfs(r, c):
        # 1. Base Case: Check bounds and visited status
        if (r < 0 or c < 0 or r == ROWS or c == COLS or
            (r, c) in visited or grid[r][c] == "0"):
            return 0
        
        # 2. Process cell
        visited.add((r, c))
        
        # 3. Traverse neighbors (Up, Down, Left, Right)
        dfs(r + 1, c)
        dfs(r - 1, c)
        dfs(r, c + 1)
        dfs(r, c - 1)
        
        return 1

    # Loop through every cell to start DFS
    count = 0
    for r in range(ROWS):
        for c in range(COLS):
            if grid[r][c] == "1" and (r, c) not in visited:
                count += dfs(r, c)
    return count
```
