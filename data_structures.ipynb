# Data Structures and Algorithm in Python
This notebook provides an overview of fundamental data structures and algorithms in Python, focusing on their time complexities. Understanding these concepts is essential for writing efficient code and optimizing performance in real-world applications.

# Operations on Data Structures

- **Traversing:** Visiting each element of data structures in order to perform some specific operation like searching or sorting.
- **Insertion:** Process of adding elements to the data structure at any location.
- **Deletion:**  Process removing an element from the data structure is called deletion.
- **Searching:** Process of finding location of an element within data structures is called searching.
- **Merging:**  Merging of two lists A & B and final merged list C of size (M+N), then this process is called merging.

# Algorithms

# Algorithm Complexities
### Time Complexity

Amount of time required to complete the execution. The time complexity of an algorithm.

We are measuring the number of operations. 

### Space Complexity

Amount of space required to solve a problem and produce an output. Similar to the time complexity, space complexity is also expressed in Big O notation.

Space Complexity = Auxiliary Space + Input Size
## Asymptotic Analysis

- Worst Case ⇒ Big - O - Notation (**O**)
- Average Case ⇒ Theta Notation (**θ**)
- Best Case ⇒ Omega Notation (**Ω**)

### Big - O - Notation (**O**)

- The Upper bound of an algorithm running time in the worst - case scenario. It represents an upper limit on the growth rate of a function.
- This is mainly used for algorithm analysis.
- `f(n)` is said to be `O(g(n))` , where `g(n)` is another function, `f(n)` gross no faster than `g(n)` for sufficient large values of `n`.
- $O(n^2)$

### Omega Notation (**Ω**)

- The lower bound of an algorithm is running time in the best case scenario. Indicate lower limit of growth rate.
- `f(n)` is `Ω(g(n)` means that `f(n)` grows at least as fast as `g(n)` for sufficiently large values of `n`.

### Theta Notation (**θ**)

- Combine both upper & lower bounds to provide a tight asymptotic symbol bound on an algorithm’s running time.
- `f(n)` is `θ(g(n))`, means that `f(n)` grows at the same rate as `g(n)` for sufficiently large value of `n`.

# Static Linear Data Structure
## Arrays (Concept)

- **Definition**: Collection of elements of the same data type stored in contiguous memory locations.
- **Properties**:
    - Fixed size.
    - Elements are accessed by **index** (0-based).
    - Supports operations: traversal, insertion, deletion, search.
- **Advantages**:
    - Fast random access → O(1).
- **Disadvantages**:
    - Fixed size → cannot resize easily.
    - Insertion/Deletion in middle is costly → O(n).

````python
from array import array

# Create integer array
arr = array('i', [10, 20, 30, 40, 50])

# Access
print(arr[2])   # 30

# Insertion
arr.insert(2, 25)  # [10, 20, 25, 30, 40, 50]

# Deletion
arr.remove(40)     # [10, 20, 25, 30, 50]

# Traversal
for item in arr:
    print(item)

# Searching
print(arr.index(30))  # 3
````

## Lists (Python’s Built-in Dynamic Array)

- **Definition**: A built-in Python data structure, more flexible than arrays.
- **Properties**:
    - Can hold mixed data types.
    - Automatically resizes (dynamic).
    - Supports nesting (lists within lists).
- **Advantages**:
    - Very flexible, rich set of methods.
    - Easy to use for most problems.
- **Disadvantages**:
    - Slightly slower than low-level arrays due to dynamic nature.

````python
# Creating lists
my_list = [1, 2, 3, 4]
mixed_list = [10, "hello", 3.14, [1, 2]]

# Access
print(my_list[1])   # 2

# Insertion
my_list.append(5)   # [1, 2, 3, 4, 5]
my_list.insert(2, 99) # [1, 2, 99, 3, 4, 5]

# Deletion
my_list.remove(99)   # [1, 2, 3, 4, 5]
my_list.pop()        # removes last → [1, 2, 3, 4]

# Traversal
for x in my_list:
    print(x)

# Searching
print(3 in my_list)    # True
print(my_list.index(3)) # 2
````

## Dictionaries

### Concept

- **Definition**: A collection of key–value pairs.
- **Properties**:
    - Unordered (in Python < 3.7, insertion order was not preserved; from 3.7+, it preserves order).
    - Keys must be **unique** and **hashable** (immutable types: int, str, tuple).
    - Values can be of any type.
    - Very efficient lookups.

````python
# Creating a dictionary
student = {"name": "Alice", "age": 21, "major": "CS"}

# Access
print(student["name"])   # Alice
print(student.get("age")) # 21

# Insertion
student["gpa"] = 3.9

# Update
student["age"] = 22

# Deletion
del student["major"]
removed = student.pop("gpa")

# Traversal
for key, value in student.items():
    print(key, ":", value)

# Searching
print("name" in student)  # True
````

### Time Complexity (Avg Case)

| Operation | Time Complexity |
| --- | --- |
| Access by key | O(1) |
| Insertion | O(1) |
| Deletion | O(1) |
| Search | O(1) |

### Use Cases

- Fast lookups (phone book, caching, routing tables).
- Storing structured data.
- Frequency counting (word count, histogram).

## Tuples

### Concept

- **Definition**: An **immutable** ordered collection of elements.
- **Properties**:
    - Similar to lists, but **cannot be changed** after creation.
    - Can contain different data types.
    - Hashable → can be used as dictionary keys.
    - Lightweight compared to lists.
⚠️ Since tuples are immutable:

- No `.append()` / `.remove()` / `.insert()`.
- Elements can’t be modified directly.

````python
# Creating tuples
t1 = (1, 2, 3)
t2 = ("apple", "banana", "cherry")
nested = (1, [2, 3], (4, 5))

# Access
print(t1[0])     # 1
print(t2[-1])    # cherry

# Traversal
for item in t1:
    print(item)

# Concatenation
t3 = t1 + (4, 5)    # (1, 2, 3, 4, 5)

# Repetition
print(t1 * 2)       # (1, 2, 3, 1, 2, 3)

# Membership test
print(2 in t1)      # True
````

### Time Complexity

| Operation | Time Complexity |
| --- | --- |
| Access by index | O(1) |
| Traversal | O(n) |
| Search | O(n) |

### Use Cases

- Protect data from modification (constants).
- As dictionary keys (immutable).
- Return multiple values from a function.
- Used in sets (because hashable).

# Dynamic Linear Data Structure

## Singly Linked List

A **Singly Linked List** is a linear data structure composed of a sequence of **nodes**. Each node contains **data** and a **pointer** (`next`) to the next node. The list is accessed through a **head** pointer, and the last node's `next` pointer is `null`.

### Node Structure

```
class Node {
  value // The data stored
  Node next // Pointer to the next node
}
```

-----

### Key Characteristics

  * ✅ **Dynamic Size**: Can easily grow or shrink at runtime.
  * ✅ **Efficient Insertions/Deletions**: Operations at the beginning are very fast ($O(1)$).
  * ❌ **No Random Access**: Accessing the $i$-th element takes $O(i)$ time as it requires traversal from the `head`.
  * ❌ **Extra Memory**: Requires extra memory for the `next` pointer in each node.

-----

### Algorithms for Common Operations 🧑‍💻

Here are the pseudocode implementations for the most common linked list operations.

#### Traversal

To visit every node in the list, usually for displaying its contents.

  * **Time Complexity**: $O(n)$

<!-- end list -->

```
function traverse(head):
  current = head
  while current is not null:
    print(current.value)
    current = current.next
```

#### Search

To find a node with a specific value.

  * **Time Complexity**: $O(n)$

<!-- end list -->

```
function search(head, key):
  current = head
  while current is not null:
    if current.value == key:
      return true // Found the key
    current = current.next
  return false // Key not found
```

#### Insertion at the Head (Prepend)

Adds a new node to the very beginning of the list.

  * **Time Complexity**: $O(1)$

<!-- end list -->

```
function prepend(head, value):
  newNode = new Node(value)
  newNode.next = head
  head = newNode
  return head
```

#### Insertion at the Tail (Append)

Adds a new node to the very end of the list.

  * **Time Complexity**: $O(n)$

<!-- end list -->

```
function append(head, value):
  newNode = new Node(value)
  
  // If the list is empty, the new node becomes the head
  if head is null:
    head = newNode
    return head
    
  // Traverse to the last node
  last = head
  while last.next is not null:
    last = last.next
    
  // Link the last node to the new node
  last.next = newNode
  return head
```

#### Deletion from the Head

Removes the first node in the list.

  * **Time Complexity**: $O(1)$

<!-- end list -->

```
function deleteFromHead(head):
  // If list is empty, do nothing
  if head is null:
    return null
    
  // The new head is the second node
  newHead = head.next
  return newHead
```

#### Deletion from the Tail

Removes the last node in the list.

  * **Time Complexity**: $O(n)$

<!-- end list -->

```
function deleteFromTail(head):
  // If list is empty or has only one node
  if head is null or head.next is null:
    return null
    
  // Traverse to the second-to-last node
  second_last = head
  while second_last.next.next is not null:
    second_last = second_last.next
    
  // Unlink the last node
  second_last.next = null
  return head
```

#### Deletion by Value

Removes the first node that contains a specific value.

  * **Time Complexity**: $O(n)$

<!-- end list -->

```
function deleteByValue(head, key):
  if head is null: return null
  
  // Case 1: The head node itself holds the key
  if head.value == key:
    return head.next
    
  // Search for the node *before* the one to be deleted
  current = head
  while current.next is not null and current.next.value != key:
    current = current.next
    
  // If the key was found, unlink the node
  if current.next is not null:
    current.next = current.next.next
    
  return head
```


````python
class SLLNode:
    def __init__(self,value):
        self.value = value
        self.next = None
````

````python
class SinglyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
        self.length = 0

    def append(self, value):
        new_node = SLLNode(value=value)
        if self.head is None:
            self.head = new_node
            self.tail = new_node
        else:
            self.tail.next = new_node
            self.tail = new_node
        self.length += 1

    def __str__(self):
        temp_node = self.head
        result = ""
        while temp_node is not None:
            result += str(temp_node.value)
            if temp_node.next is not None:
                result += " -> "
            temp_node = temp_node.next
        return result
````
