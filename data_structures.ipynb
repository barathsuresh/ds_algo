# Data Structures and Algorithm in Python

This notebook provides an overview of fundamental data structures and algorithms in Python, focusing on their time complexities. Understanding these concepts is essential for writing efficient code and optimizing performance in real-world applications.


# Operations on Data Structures

- **Traversing:** Visiting each element of data structures in order to perform some specific operation like searching or sorting.
- **Insertion:** Process of adding elements to the data structure at any location.
- **Deletion:** Process removing an element from the data structure is called deletion.
- **Searching:** Process of finding location of an element within data structures is called searching.
- **Merging:** Merging of two lists A & B and final merged list C of size (M+N), then this process is called merging.


# Algorithms

An algorithm is a process or set of rules required to perform calculations or some other problem - solving operations especially by a computer.

## Approaches in Algorithm

### Brute force Algorithm

exhaustive search algorithm that searches all possible to provide require a solution.

- **Optimizing:** Finding all solution of a problem and then take out of best solution.
- **Sacrificing:** As soon as the best solution is found then it will stop.

### Divide and Conquer Algorithm

Breaks down the algorithm solve the problem.

### Greedy Algorithm

Makes an optimal choice on each iteration with the hope of getting best solution.

## Major Categories of Algorithms

- Sort
- Search
- Delete
- Insert
- Update


# Algorithm Complexities

### Time Complexity

Amount of time required to complete the execution. The time complexity of an algorithm.

We are measuring the number of operations.

### Space Complexity

Amount of space required to solve a problem and produce an output. Similar to the time complexity, space complexity is also expressed in Big O notation.

Space Complexity = Auxiliary Space + Input Size

## Asymptotic Analysis

- Worst Case ⇒ Big - O - Notation (**O**)
- Average Case ⇒ Theta Notation (**θ**)
- Best Case ⇒ Omega Notation (**Ω**)

### Big - O - Notation (**O**)

- The Upper bound of an algorithm running time in the worst - case scenario. It represents an upper limit on the growth rate of a function.
- This is mainly used for algorithm analysis.
- `f(n)` is said to be `O(g(n))` , where `g(n)` is another function, `f(n)` gross no faster than `g(n)` for sufficient large values of `n`.
- $O(n^2)$

### Omega Notation (**Ω**)

- The lower bound of an algorithm is running time in the best case scenario. Indicate lower limit of growth rate.
- `f(n)` is `Ω(g(n)` means that `f(n)` grows at least as fast as `g(n)` for sufficiently large values of `n`.

### Theta Notation (**θ**)

- Combine both upper & lower bounds to provide a tight asymptotic symbol bound on an algorithm’s running time.
- `f(n)` is `θ(g(n))`, means that `f(n)` grows at the same rate as `g(n)` for sufficiently large value of `n`.

# Static Linear Data Structure

## Arrays (Concept)

- **Definition**: Collection of elements of the same data type stored in contiguous memory locations.
- **Properties**:
  - Fixed size.
  - Elements are accessed by **index** (0-based).
  - Supports operations: traversal, insertion, deletion, search.
- **Advantages**:
  - Fast random access → O(1).
- **Disadvantages**:
  - Fixed size → cannot resize easily.
  - Insertion/Deletion in middle is costly → O(n).


````python
from array import array

# Create integer array
arr = array("i", [10, 20, 30, 40, 50])

# Access
print(arr[2])  # 30

# Insertion
arr.insert(2, 25)  # [10, 20, 25, 30, 40, 50]

# Deletion
arr.remove(40)  # [10, 20, 25, 30, 50]

# Traversal
for item in arr:
    print(item)

# Searching
print(arr.index(30))  # 3
````

## Lists (Python’s Built-in Dynamic Array)

- **Definition**: A built-in Python data structure, more flexible than arrays.
- **Properties**:
  - Can hold mixed data types.
  - Automatically resizes (dynamic).
  - Supports nesting (lists within lists).
- **Advantages**:
  - Very flexible, rich set of methods.
  - Easy to use for most problems.
- **Disadvantages**:
  - Slightly slower than low-level arrays due to dynamic nature.


````python
# Creating lists
my_list = [1, 2, 3, 4]
mixed_list = [10, "hello", 3.14, [1, 2]]

# Access
print(my_list[1])  # 2

# Insertion
my_list.append(5)  # [1, 2, 3, 4, 5]
my_list.insert(2, 99)  # [1, 2, 99, 3, 4, 5]

# Deletion
my_list.remove(99)  # [1, 2, 3, 4, 5]
my_list.pop()  # removes last → [1, 2, 3, 4]

# Traversal
for x in my_list:
    print(x)

# Searching
print(3 in my_list)  # True
print(my_list.index(3))  # 2
````

## Singly Linked List

A **Singly Linked List** is a linear data structure composed of a sequence of **nodes**. Each node contains **data** and a **pointer** (`next`) to the next node. The list is accessed through a **head** pointer, and the last node's `next` pointer is `null`.

### Node Structure

```
class Node {
  value // The data stored
  Node next // Pointer to the next node
}
```

---

### Key Characteristics

- ✅ **Dynamic Size**: Can easily grow or shrink at runtime.
- ✅ **Efficient Insertions/Deletions**: Operations at the beginning are very fast ($O(1)$).
- ❌ **No Random Access**: Accessing the $i$-th element takes $O(i)$ time as it requires traversal from the `head`.
- ❌ **Extra Memory**: Requires extra memory for the `next` pointer in each node.

---

### Algorithms for Common Operations 🧑‍💻

Here are the pseudocode implementations for the most common linked list operations.

#### Traversal

To visit every node in the list, usually for displaying its contents.

- **Time Complexity**: $O(n)$

<!-- end list -->

```
function traverse(head):
  current = head
  while current is not null:
    print(current.value)
    current = current.next
```

#### Search

To find a node with a specific value.

- **Time Complexity**: $O(n)$

<!-- end list -->

```
function search(head, key):
  current = head
  while current is not null:
    if current.value == key:
      return true // Found the key
    current = current.next
  return false // Key not found
```

#### Insertion at the Head (Prepend)

Adds a new node to the very beginning of the list.

- **Time Complexity**: $O(1)$

<!-- end list -->

```
function prepend(head, value):
  newNode = new Node(value)
  newNode.next = head
  head = newNode
  return head
```

#### Insertion at the Tail (Append)

Adds a new node to the very end of the list.

- **Time Complexity**: $O(n)$

<!-- end list -->

```
function append(head, value):
  newNode = new Node(value)

  // If the list is empty, the new node becomes the head
  if head is null:
    head = newNode
    return head

  // Traverse to the last node
  last = head
  while last.next is not null:
    last = last.next

  // Link the last node to the new node
  last.next = newNode
  return head
```

#### Deletion from the Head

Removes the first node in the list.

- **Time Complexity**: $O(1)$

<!-- end list -->

```
function deleteFromHead(head):
  // If list is empty, do nothing
  if head is null:
    return null

  // The new head is the second node
  newHead = head.next
  return newHead
```

#### Deletion from the Tail

Removes the last node in the list.

- **Time Complexity**: $O(n)$

<!-- end list -->

```
function deleteFromTail(head):
  // If list is empty or has only one node
  if head is null or head.next is null:
    return null

  // Traverse to the second-to-last node
  second_last = head
  while second_last.next.next is not null:
    second_last = second_last.next

  // Unlink the last node
  second_last.next = null
  return head
```

#### Deletion by Value

Removes the first node that contains a specific value.

- **Time Complexity**: $O(n)$

<!-- end list -->

```
function deleteByValue(head, key):
  if head is null: return null

  // Case 1: The head node itself holds the key
  if head.value == key:
    return head.next

  // Search for the node *before* the one to be deleted
  current = head
  while current.next is not null and current.next.value != key:
    current = current.next

  // If the key was found, unlink the node
  if current.next is not null:
    current.next = current.next.next

  return head
```
| Operation | Time | Space | Notes |
|---|---|---|---|
| `is_empty` | O(1) | O(1) | size counter |
| `__len__`  | O(1) | O(1) | size counter |
| `clear`    | O(1) | O(1) | let GC reclaim nodes |
| `prepend`  | O(1) | O(1) | insert at head |
| `append`   | O(n) | O(1) | walk to tail |
| `insert_at(i)` | O(n) | O(1) | walk to position |
| `delete_at(i)` | O(n) | O(1) | walk to position |
| `delete_value(v)` | O(n) | O(1) | scan until match |
| `find(v)` | O(n) | O(1) | linear search |
| `get(i)` / `set(i,v)` | O(n) | O(1) | walk to i |
| `reverse()` | O(n) | O(1) | pointer reversal |
| `to_list()` | O(n) | O(1) | gather values |


````python
class SLLNode:
    __slots__ = ("data", "next")
    def __init__(self, data, next=None):
        self.data = data
        self.next = next

class SinglyLinkedList:
    def __init__(self, iterable=None):
        self.head = None
        self._size = 0
        if iterable:
            for x in iterable: self.append(x)

    def is_empty(self): return self._size == 0
    def __len__(self): return self._size
    def clear(self): self.head=None; self._size=0
    def __iter__(self):
        cur = self.head
        while cur: 
            yield cur.data; cur = cur.next
    def __repr__(self):
        return f"SinglyLinkedList([{', '.join(repr(x) for x in self)}])"
    def _check_index(self, index, allow_end=False):
        if index < 0 or (index > self._size if allow_end else index >= self._size):
            raise IndexError("index out of range")

    def prepend(self, value):
        self.head = SLLNode(value, self.head); self._size += 1

    def append(self, value):
        node = SLLNode(value)
        if not self.head: self.head = node
        else:
            cur = self.head
            while cur.next: cur = cur.next
            cur.next = node
        self._size += 1

    def insert_at(self, index, value):
        self._check_index(index, allow_end=True)
        if index == 0: self.prepend(value); return
        prev = self.head
        for _ in range(index-1): prev = prev.next
        prev.next = SLLNode(value, prev.next)
        self._size += 1

    def delete_at(self, index):
        self._check_index(index)
        prev, cur = None, self.head
        for _ in range(index): prev, cur = cur, cur.next
        if prev is None: self.head = cur.next
        else: prev.next = cur.next
        self._size -= 1
        return cur.data

    def delete_value(self, value):
        prev, cur = None, self.head
        while cur:
            if cur.data == value:
                if prev is None: self.head = cur.next
                else: prev.next = cur.next
                self._size -= 1; return True
            prev, cur = cur, cur.next
        return False

    def find(self, value):
        cur, i = self.head, 0
        while cur:
            if cur.data == value: return i
            cur, i = cur.next, i+1
        return -1

    def get(self, index):
        self._check_index(index)
        cur = self.head
        for _ in range(index): cur = cur.next
        return cur.data

    def set(self, index, value):
        self._check_index(index)
        cur = self.head
        for _ in range(index): cur = cur.next
        cur.data = value

    def to_list(self): return list(iter(self))

    def reverse(self):
        prev, cur = None, self.head
        while cur:
            nxt = cur.next
            cur.next = prev
            prev, cur = cur, nxt
        self.head = prev
    
    # ---------- Pretty printer (no class changes needed) ----------
def as_arrow(lst):
    """Return list in '1 -> 2 -> 3' form; 'Empty' if no nodes."""
    return "Empty" if lst.is_empty() else " -> ".join(str(x) for x in lst)

def show(stage, lst):
    """Print a stage label + current list + size."""
    print(f"{stage:<30} | {as_arrow(lst)}  (len={len(lst)})")

# ---------- Test Suite for SinglyLinkedList ----------
def run_sll_tests():
    print("=== SinglyLinkedList Tests ===")

    # Build from iterable
    s = SinglyLinkedList([1, 2, 3])
    show("Init from [1,2,3]", s)
    assert len(s) == 3 and s.to_list() == [1, 2, 3]

    # prepend
    s.prepend(0)
    show("After prepend(0)", s)
    assert s.to_list() == [0, 1, 2, 3]

    # append
    s.append(4)
    show("After append(4)", s)
    assert s.to_list() == [0, 1, 2, 3, 4]

    # insert_at middle
    s.insert_at(2, 99)  # [0,1,99,2,3,4]
    show("After insert_at(2,99)", s)
    assert s.to_list() == [0, 1, 99, 2, 3, 4]

    # insert_at at head (index 0)
    s.insert_at(0, -1)  # [-1,0,1,99,2,3,4]
    show("After insert_at(0,-1)", s)
    assert s.to_list() == [-1, 0, 1, 99, 2, 3, 4]

    # insert_at at end (index == len)
    s.insert_at(len(s), 5)  # [-1,0,1,99,2,3,4,5]
    show("After insert_at(len,5)", s)
    assert s.to_list() == [-1, 0, 1, 99, 2, 3, 4, 5]

    # find existing and non-existing
    idx_99 = s.find(99)
    idx_100 = s.find(100)
    print(f"find(99) -> {idx_99}, find(100) -> {idx_100}")
    assert idx_99 == 3 and idx_100 == -1

    # get / set
    val_at_3 = s.get(3)   # should be 99
    print(f"get(3) -> {val_at_3}")
    s.set(3, 42)          # replace 99 with 42
    show("After set(3,42)", s)
    assert s.to_list() == [-1, 0, 1, 42, 2, 3, 4, 5]

    # delete_at head
    removed = s.delete_at(0)  # removes -1
    print(f"delete_at(0) -> {removed}")
    show("After delete_at(0)", s)
    assert removed == -1 and s.to_list() == [0, 1, 42, 2, 3, 4, 5]

    # delete_at middle
    removed = s.delete_at(2)  # removes 42
    print(f"delete_at(2) -> {removed}")
    show("After delete_at(2)", s)
    assert removed == 42 and s.to_list() == [0, 1, 2, 3, 4, 5]

    # delete_at tail
    removed = s.delete_at(len(s) - 1)  # removes 5
    print(f"delete_at(len-1) -> {removed}")
    show("After delete_at(len-1)", s)
    assert removed == 5 and s.to_list() == [0, 1, 2, 3, 4]

    # delete_value existing (first occurrence)
    ok = s.delete_value(2)
    print(f"delete_value(2) -> {ok}")
    show("After delete_value(2)", s)
    assert ok is True and s.to_list() == [0, 1, 3, 4]

    # delete_value missing
    ok = s.delete_value(999)
    print(f"delete_value(999) -> {ok}")
    show("Attempt delete_value(999)", s)
    assert ok is False and s.to_list() == [0, 1, 3, 4]

    # reverse
    s.reverse()
    show("After reverse()", s)
    assert s.to_list() == [4, 3, 1, 0]

    # clear
    s.clear()
    show("After clear()", s)
    assert s.is_empty() and s.to_list() == []

    # edge cases: errors on invalid operations
    try:
        s.delete_at(0)
    except IndexError as e:
        print(f"delete_at(0) on empty -> IndexError: {e}")

    try:
        s.get(0)
    except IndexError as e:
        print(f"get(0) on empty -> IndexError: {e}")

    # rebuild quickly to verify append/prepend combined
    s.append(10); show("append(10)", s)
    s.prepend(9); show("prepend(9)", s)
    s.append(11); show("append(11)", s)
    assert s.to_list() == [9, 10, 11]

    print("\n✅ All SLL tests passed!")
````

````python
run_sll_tests()
````

## Doubly Linked List

A **Doubly Linked List (DLL)** is a linear data structure where each node stores **data**, a pointer to the **next** node, and a pointer to the **previous** node. We typically keep pointers to both **head** and **tail** to support O(1) insertion/removal at the ends.

### Node Structure
```
class Node {
  value       // data stored
  Node prev   // pointer to the previous node
  Node next   // pointer to the next node
}
```
---
### Key Characteristics
- ✅ **Bidirectional traversal** using `prev` and `next`.
- ✅ **O(1) insert/delete at both ends** with head/tail pointers.
- ❌ **Extra memory** for the `prev` pointer in each node.
- ❌ Slightly more complex pointer maintenance during insert/delete.



### Algorithms for Common Operations 🧑‍💻

#### Traversal — O(n)
```
function traverse(head):
  current = head
  while current != null:
    print(current.value)
    current = current.next
```
#### Search — O(n)
```
function search(head, key):
  current = head
  while current != null:
    if current.value == key: return true
    current = current.next
  return false
```
#### Prepend — O(1)
```
function prepend(head, tail, value):
  new = Node(value)
  new.next = head
  new.prev = null
  if head != null: head.prev = new else: tail = new
  head = new
```
#### Append — O(1)
```
function append(head, tail, value):
  new = Node(value)
  new.prev = tail
  new.next = null
  if tail != null: tail.next = new else: head = new
  tail = new
```
#### Delete by Value — O(n)
```
function deleteByValue(head, tail, key):
  cur = head
  while cur != null and cur.value != key: cur = cur.next
  if cur == null: return (head, tail)
  if cur.prev != null: cur.prev.next = cur.next else: head = cur.next
  if cur.next != null: cur.next.prev = cur.prev else: tail = cur.prev
```
| Operation | Time | Space | Notes |
|---|---|---|---|
| `is_empty` | O(1) | O(1) | size counter |
| `__len__` | O(1) | O(1) | size counter |
| `clear` | O(1) | O(1) | drop references |
| `prepend` | O(1) | O(1) | use head |
| `append` | O(1) | O(1) | use tail |
| `insert_at(i)` | O(n) | O(1) | walk from nearer end |
| `delete_at(i)` | O(n) | O(1) | walk from nearer end |
| `delete_value(v)` | O(n) | O(1) | scan |
| `find(v)` | O(n) | O(1) | linear search |
| `get/set(i)` | O(n) | O(1) | walk from nearer end |
| `reverse()` | O(n) | O(1) | swap prev/next |
| `to_list()` | O(n) | O(1) | gather values |


````python
# ========================
# Doubly Linked List (DLL)
# ========================

class DLLNode:
    __slots__ = ("data","prev","next")                         # Save memory by fixing attributes
    def __init__(self, data, prev=None, next=None):            # Initialize a new doubly-linked node
        self.data = data                                       # Store the payload value
        self.prev = prev                                       # Pointer to previous node (None if first)
        self.next = next                                       # Pointer to next node (None if last)

class DoublyLinkedList:
    def __init__(self, iterable=None):                         # Optionally build list from an iterable
        self.head = None                                       # Reference to first node (None if empty)
        self.tail = None                                       # Reference to last node (None if empty)
        self._size = 0                                         # Keep a size counter for O(1) length
        if iterable is not None:                               # If we got initial values
            for x in iterable:                                 # Iterate through the provided elements
                self.append(x)                                 # Append each element to the tail (O(1) per op)

    def __repr__(self):                                        # Developer-friendly string representation
        cur = self.head                                        # Start traversal at head
        vals = []                                              # Accumulate stringified payloads
        while cur is not None:                                 # Traverse until the end
            vals.append(repr(cur.data))                        # Add the current node's data representation
            cur = cur.next                                     # Move to next node
        return f"DoublyLinkedList([{', '.join(vals)}])"        # Show list in Python list-like form

    def __len__(self):                                         # Make len(dll) work in O(1)
        return self._size                                      # Return stored size

    def is_empty(self):                                        # Check emptiness in O(1)
        return self._size == 0                                 # True when size is zero

    def clear(self):                                           # Remove all elements quickly
        self.head = self.tail = None                           # Drop references so GC can reclaim nodes
        self._size = 0                                         # Reset size counter to 0

    def prepend(self, value):                                  # Insert new node at the head in O(1)
        node = DLLNode(value, None, self.head)                 # New node's next points to old head
        if self.head is not None:                              # If list was non-empty
            self.head.prev = node                              # Old head's prev should point back to new node
        else:                                                  # If list was empty
            self.tail = node                                   # New node is also the tail
        self.head = node                                       # Update head to the new node
        self._size += 1                                        # Increment size due to insertion

    def append(self, value):                                   # Insert new node at the tail in O(1)
        node = DLLNode(value, self.tail, None)                 # New node's prev points to old tail
        if self.tail is not None:                              # If list was non-empty
            self.tail.next = node                              # Old tail should point forward to new node
        else:                                                  # If list was empty
            self.head = node                                   # New node becomes the head
        self.tail = node                                       # Update tail to the new node
        self._size += 1                                        # Increment size counter

    def _check_index(self, i, allow_end=False):                # Validate index for get/set/insert/delete
        if i < 0 or (i > self._size if allow_end else i >= self._size):  # Out-of-bounds check
            raise IndexError("index out of range")             # Raise Pythonic error

    def _node_at(self, i):                                     # Return node at index i (walk from nearer end)
        self._check_index(i)                                   # Ensure index is valid
        if i <= self._size // 2:                               # If i is in the first half
            cur = self.head                                    # Start from head
            for _ in range(i):                                 # Walk i steps forward
                cur = cur.next                                 # Advance to next node
        else:                                                  # Else i is in the second half
            cur = self.tail                                    # Start from tail
            for _ in range(self._size - 1 - i):                # Walk backward the remaining steps
                cur = cur.prev                                 # Move to previous node
        return cur                                             # Return the located node

    def insert_at(self, i, value):                             # Insert value at position i (0..size)
        self._check_index(i, allow_end=True)                   # Allow insertion at end (i == size)
        if i == 0:                                             # If inserting at head
            self.prepend(value)                                # Delegate to O(1) prepend
            return                                             # Done
        if i == self._size:                                    # If inserting at tail
            self.append(value)                                 # Delegate to O(1) append
            return                                             # Done
        cur = self._node_at(i)                                 # Node currently at index i
        node = DLLNode(value, cur.prev, cur)                   # New node placed before cur
        cur.prev.next = node                                   # Link previous neighbor forward to new node
        cur.prev = node                                        # Link current node back to new node
        self._size += 1                                        # Bump size

    def delete_at(self, i):                                    # Delete node at index i and return its value
        cur = self._node_at(i)                                 # Locate node to delete
        val = cur.data                                         # Keep value to return
        if cur.prev is not None:                               # If there is a previous node
            cur.prev.next = cur.next                           # Bridge previous to next (unlink cur)
        else:                                                  # If cur was head
            self.head = cur.next                               # Move head forward
        if cur.next is not None:                               # If there is a next node
            cur.next.prev = cur.prev                           # Bridge next to previous (unlink cur)
        else:                                                  # If cur was tail
            self.tail = cur.prev                               # Move tail backward
        self._size -= 1                                        # Decrement size
        return val                                             # Return deleted value

    def delete_value(self, value):                             # Delete first occurrence of value
        cur = self.head                                        # Start at head
        while cur is not None:                                 # Traverse until end
            if cur.data == value:                              # Found target value
                if cur.prev is not None:                       # Fix prev->next if there is a previous node
                    cur.prev.next = cur.next                   # Bypass cur
                else:                                          # cur is head
                    self.head = cur.next                       # Move head forward
                if cur.next is not None:                       # Fix next->prev if there is a next node
                    cur.next.prev = cur.prev                   # Bypass cur
                else:                                          # cur is tail
                    self.tail = cur.prev                       # Move tail backward
                self._size -= 1                                # Decrement size
                return True                                    # Signal deletion happened
            cur = cur.next                                     # Continue scanning
        return False                                           # Value not found

    def find(self, value):                                     # Return index of first occurrence or -1
        cur, i = self.head, 0                                  # Start at head with index counter
        while cur is not None:                                 # Traverse until end
            if cur.data == value:                              # If value matches
                return i                                       # Return current index
            cur, i = cur.next, i + 1                           # Step forward and increment index
        return -1                                              # Not found

    def get(self, i):                                          # Random read by index (O(n))
        return self._node_at(i).data                           # Locate node then return its data

    def set(self, i, value):                                   # Overwrite value at index (O(n))
        self._node_at(i).data = value                          # Locate node then assign new data

    def to_list(self):                                         # Convert to plain Python list
        out, cur = [], self.head                               # Prepare output array and start pointer
        while cur is not None:                                 # Traverse nodes
            out.append(cur.data)                               # Append each data to result list
            cur = cur.next                                     # Advance pointer
        return out                                             # Return the built list

    def reverse(self):                                         # Reverse the list in-place (O(n), O(1) space)
        cur = self.head                                        # Start from current head
        self.tail = cur if cur is not None else None           # Track new tail (old head)
        prev_node = None                                       # Previous node for pointer flipping
        while cur is not None:                                 # Walk the list once
            nxt = cur.next                                     # Store next before we overwrite it
            cur.next = prev_node                               # Flip next pointer to point backward
            cur.prev = nxt                                     # Update prev to old next (consistent DLL links)
            prev_node = cur                                    # Advance prev_node forward
            cur = nxt                                          # Move to original next
        self.head = prev_node                                  # After loop, prev_node is new head

# ---------- Pretty printer for DLL (bidirectional view) ----------
def as_double_arrow(dll):
    """Return a bidirectional view like: None <- 1 <-> 2 <-> 3 -> None."""
    if dll.is_empty():
        return "Empty"
    values = " <-> ".join(str(x) for x in dll.to_list())  # join with <-> for DLL
    return f"None <- {values} -> None"                    # show both ends as None

def show_dll(stage, dll):
    """Print stage label + current DLL + size."""
    print(f"{stage:<32} | {as_double_arrow(dll)}  (len={len(dll)})")

# ---------- Test Suite for DoublyLinkedList ----------
def run_dll_tests():
    print("=== DoublyLinkedList Tests ===")

    # Init from iterable
    dll = DoublyLinkedList([1, 2, 3])
    show_dll("Init from [1,2,3]", dll)
    assert len(dll) == 3 and dll.to_list() == [1, 2, 3]

    # prepend
    dll.prepend(0)  # [0,1,2,3]
    show_dll("After prepend(0)", dll)
    assert dll.to_list() == [0, 1, 2, 3]

    # append
    dll.append(4)  # [0,1,2,3,4]
    show_dll("After append(4)", dll)
    assert dll.to_list() == [0, 1, 2, 3, 4]

    # insert_at middle
    dll.insert_at(2, 99)  # [0,1,99,2,3,4]
    show_dll("After insert_at(2,99)", dll)
    assert dll.to_list() == [0, 1, 99, 2, 3, 4]

    # insert_at head (0)
    dll.insert_at(0, -1)  # [-1,0,1,99,2,3,4]
    show_dll("After insert_at(0,-1)", dll)
    assert dll.to_list() == [-1, 0, 1, 99, 2, 3, 4]

    # insert_at end (len)
    dll.insert_at(len(dll), 5)  # [-1,0,1,99,2,3,4,5]
    show_dll("After insert_at(len,5)", dll)
    assert dll.to_list() == [-1, 0, 1, 99, 2, 3, 4, 5]

    # find existing & non-existing
    idx_99 = dll.find(99)
    idx_100 = dll.find(100)
    print(f"find(99) -> {idx_99}, find(100) -> {idx_100}")
    assert idx_99 == 3 and idx_100 == -1

    # get / set
    v3 = dll.get(3)  # currently 99
    print(f"get(3) -> {v3}")
    dll.set(3, 42)   # replace 99 with 42
    show_dll("After set(3,42)", dll)
    assert dll.to_list() == [-1, 0, 1, 42, 2, 3, 4, 5]

    # delete_at head
    removed = dll.delete_at(0)  # remove -1
    print(f"delete_at(0) -> {removed}")
    show_dll("After delete_at(0)", dll)
    assert removed == -1 and dll.to_list() == [0, 1, 42, 2, 3, 4, 5]

    # delete_at middle
    removed = dll.delete_at(2)  # remove 42
    print(f"delete_at(2) -> {removed}")
    show_dll("After delete_at(2)", dll)
    assert removed == 42 and dll.to_list() == [0, 1, 2, 3, 4, 5]

    # delete_at tail
    removed = dll.delete_at(len(dll) - 1)  # remove 5
    print(f"delete_at(len-1) -> {removed}")
    show_dll("After delete_at(len-1)", dll)
    assert removed == 5 and dll.to_list() == [0, 1, 2, 3, 4]

    # delete_value existing (first occurrence)
    ok = dll.delete_value(2)
    print(f"delete_value(2) -> {ok}")
    show_dll("After delete_value(2)", dll)
    assert ok is True and dll.to_list() == [0, 1, 3, 4]

    # delete_value missing
    ok = dll.delete_value(999)
    print(f"delete_value(999) -> {ok}")
    show_dll("Attempt delete_value(999)", dll)
    assert ok is False and dll.to_list() == [0, 1, 3, 4]

    # reverse
    dll.reverse()
    show_dll("After reverse()", dll)
    assert dll.to_list() == [4, 3, 1, 0]

    # clear
    dll.clear()
    show_dll("After clear()", dll)
    assert dll.is_empty() and dll.to_list() == []

    # Edge cases: invalid ops on empty
    try:
        dll.delete_at(0)
    except IndexError as e:
        print(f"delete_at(0) on empty -> IndexError: {e}")

    try:
        dll.get(0)
    except IndexError as e:
        print(f"get(0) on empty -> IndexError: {e}")

    # Rebuild quickly to verify append/prepend mixed
    dll.append(10); show_dll("append(10)", dll)
    dll.prepend(9); show_dll("prepend(9)", dll)
    dll.append(11); show_dll("append(11)", dll)
    assert dll.to_list() == [9, 10, 11]

    print("\n✅ All DLL tests passed!")
````

## Circular Singly Linked List

A **Circular Singly Linked List (CSLL)** links the last node back to the **head**, forming a ring. We maintain a **tail** pointer so that `tail.next` is always the head, enabling O(1) append and prepend.

### Node Structure
```
class Node {
  value     // data stored
  Node next // pointer to the next node (head if last node)
}
```
---
### Key Characteristics
- ✅ **Circular**: last node points back to the head (`tail.next = head`).
- ✅ **O(1) append/prepend** using a maintained tail pointer.
- ❌ Traversals must stop after one **full loop** to avoid infinite cycles.
- ❌ No random access; linear search only.

### Algorithms for Common Operations 🧑‍💻

#### Traversal — O(n)
```
function traverse(tail):
  if tail == null: return
  cur = tail.next
  repeat n times:
    print(cur.value)
    cur = cur.next
```
#### Search — O(n)
```
function search(tail, key):
  if tail == null: return false
  cur = tail.next
  repeat n times:
    if cur.value == key: return true
    cur = cur.next
  return false
```
#### Prepend — O(1)
```
function prepend(tail, value):
  new = Node(value)
  if tail == null: new.next = new; tail = new
  else: new.next = tail.next; tail.next = new
  return tail
```
#### Append — O(1)
```
function append(tail, value):
  new = Node(value)
  if tail == null: new.next = new; tail = new
  else: new.next = tail.next; tail.next = new; tail = new
  return tail
```
| Operation | Time | Space | Notes |
|---|---|---|---|
| `is_empty` | O(1) | O(1) | size counter |
| `__len__` | O(1) | O(1) | size counter |
| `clear` | O(1) | O(1) | drop tail |
| `prepend` | O(1) | O(1) | use tail.next |
| `append` | O(1) | O(1) | maintain tail |
| `insert_at(i)` | O(n) | O(1) | walk from head |
| `delete_at(i)` | O(n) | O(1) | walk with (prev,cur) |
| `delete_value(v)` | O(n) | O(1) | scan |
| `find(v)` | O(n) | O(1) | linear search |
| `get/set(i)` | O(n) | O(1) | walk to i |
| `to_list()` | O(n) | O(1) | gather values |


````python
# ==================================
# Circular Singly Linked List (CSLL)
# ==================================

class CSLLNode:
    __slots__ = ("data","next")                                  # Define fields to save memory
    def __init__(self, data, next=None):                         # Initialize node
        self.data = data                                         # Store payload
        self.next = next                                         # Next pointer (will wrap to head in ring)

class CircularSinglyLinkedList:
    def __init__(self, iterable=None):                           # Optional bulk construction
        self.tail = None                                         # Tail pointer; when not None, tail.next is head
        self._size = 0                                           # Keep size for O(1) queries
        if iterable is not None:                                 # If values provided
            for x in iterable:                                   # Iterate through them
                self.append(x)                                   # Append each one in O(1)

    def __len__(self):                                           # Length in O(1)
        return self._size                                        # Return stored size

    def is_empty(self):                                          # Emptiness test
        return self._size == 0                                   # True if no elements

    def head(self):                                              # Compute head pointer
        return None if self.tail is None else self.tail.next     # Head is tail.next when list is non-empty

    def __repr__(self):                                          # Developer-friendly repr
        return f"CircularSinglyLinkedList({self.to_list()})"    # Convert to list for readability

    def clear(self):                                             # Remove all elements
        self.tail = None                                         # Drop ring by forgetting tail
        self._size = 0                                           # Reset size

    def append(self, value):                                     # Insert at tail in O(1)
        node = CSLLNode(value)                                   # Create new node
        if not self.tail:                                        # If list is currently empty
            node.next = node                                     # Node points to itself (single-node ring)
            self.tail = node                                     # Tail becomes the new node
        else:                                                    # Non-empty ring
            node.next = self.tail.next                           # New node should point to the head
            self.tail.next = node                                # Old tail should point to new node
            self.tail = node                                     # Move tail forward to the new node
        self._size += 1                                          # Increase size counter

    def prepend(self, value):                                    # Insert at head in O(1)
        node = CSLLNode(value)                                   # Create new node
        if not self.tail:                                        # If list is empty
            node.next = node                                     # Single-node ring
            self.tail = node                                     # Tail becomes this node
        else:                                                    # Non-empty
            node.next = self.tail.next                           # New node points to current head
            self.tail.next = node                                # Tail now points to the new node (new head)
        self._size += 1                                          # Increase size

    def insert_at(self, index, value):                           # Insert at index (0..size)
        if index < 0 or index > self._size:                      # Validate index range
            raise IndexError("index out of range")               # Pythonic error
        if index == 0:                                           # Inserting at head
            self.prepend(value)                                  # Delegate to prepend (O(1))
            return                                               # Done
        if index == self._size:                                  # Inserting at end
            self.append(value)                                   # Delegate to append (O(1))
            return                                               # Done
        prev = self.tail.next                                    # Start from head
        for _ in range(index - 1):                               # Walk to node just before target slot
            prev = prev.next                                     # Advance pointer
        prev.next = CSLLNode(value, prev.next)                   # Link new node into the ring
        self._size += 1                                          # Increase size

    def delete_at(self, index):                                  # Delete node by index and return its value
        if self.is_empty():                                      # Disallow deletion from empty list
            raise IndexError("delete from empty list")           # Raise error
        if index < 0 or index >= self._size:                     # Validate index
            raise IndexError("index out of range")               # Raise error
        if self._size == 1:                                      # Single node case
            val = self.tail.data                                 # Keep value to return
            self.tail = None                                     # Break ring
            self._size = 0                                       # Reset size
            return val                                           # Return removed value
        prev = self.tail                                         # Start at tail so (prev,cur) begins at (tail, head)
        cur = self.tail.next                                     # cur is head
        for _ in range(index):                                   # Walk index steps
            prev, cur = cur, cur.next                            # Slide (prev,cur) window forward
        val = cur.data                                           # Keep value to return
        prev.next = cur.next                                     # Unlink cur from ring
        if cur is self.tail:                                     # If we removed the tail
            self.tail = prev                                     # Move tail backward
        self._size -= 1                                          # Decrease size
        return val                                               # Return removed value

    def delete_value(self, value):                               # Delete first occurrence of value
        if self.is_empty():                                      # No elements => nothing to delete
            return False                                         # Signal not found
        prev = self.tail                                         # Start with prev at tail
        cur = self.tail.next                                     # And cur at head
        for _ in range(self._size):                              # Loop for at most n steps
            if cur.data == value:                                # Found target
                if self._size == 1:                              # Only one node in ring
                    self.tail = None                             # Break ring
                else:                                            # More than one node
                    prev.next = cur.next                         # Bypass cur
                    if cur is self.tail:                         # If we deleted tail
                        self.tail = prev                         # Move tail to prev
                self._size -= 1                                  # Decrease size
                return True                                      # Deletion happened
            prev, cur = cur, cur.next                            # Advance sliding window
        return False                                             # Value not found

    def find(self, value):                                       # Return index of first match or -1
        if self.is_empty():                                      # Empty ring has nothing
            return -1                                            # Not found
        cur = self.tail.next                                     # Start at head
        for i in range(self._size):                              # Visit each node once
            if cur.data == value:                                # Match?
                return i                                         # Return index
            cur = cur.next                                       # Step forward
        return -1                                                # Not found

    def get(self, index):                                        # Read by index (O(n))
        if index < 0 or index >= self._size:                     # Validate index
            raise IndexError("index out of range")               # Error if invalid
        cur = self.tail.next                                     # Start at head
        for _ in range(index):                                   # Walk index steps
            cur = cur.next                                       # Advance
        return cur.data                                          # Return payload

    def set(self, index, value):                                 # Overwrite by index (O(n))
        if index < 0 or index >= self._size:                     # Validate index
            raise IndexError("index out of range")               # Error if invalid
        cur = self.tail.next                                     # Start at head
        for _ in range(index):                                   # Walk index steps
            cur = cur.next                                       # Advance
        cur.data = value                                         # Replace payload

    def to_list(self):                                           # Convert ring to plain list for display/testing
        if self.is_empty():                                      # Empty -> empty list
            return []                                            # Return empty Python list
        out = []                                                 # Prepare result container
        cur = self.tail.next                                     # Start at head
        for _ in range(self._size):                              # Iterate exactly n times
            out.append(cur.data)                                 # Collect value
            cur = cur.next                                       # Advance pointer
        return out                                               # Return collected list

# ---------- Pretty printer for CSLL (circular view) ----------
def as_circular_arrow(csll):
    """Render as: [H]1 -> 2 -> 3 -> (back to [H]); 'Empty' if no nodes."""
    if csll.is_empty():
        return "Empty"
    out = []
    cur = csll.tail.next            # head
    n = len(csll)
    for i in range(n):
        out.append(f"[H]{cur.data}" if i == 0 else str(cur.data))
        cur = cur.next
    return " -> ".join(out) + " -> (back to [H])"

def show_csll(stage, csll):
    """Print stage label + current CSLL + size."""
    print(f"{stage:<34} | {as_circular_arrow(csll)}  (len={len(csll)})")

# ---------- Test Suite for CircularSinglyLinkedList ----------
def run_csll_tests():
    print("=== CircularSinglyLinkedList Tests ===")

    # Init from iterable
    csll = CircularSinglyLinkedList([1, 2, 3])
    show_csll("Init from [1,2,3]", csll)
    assert len(csll) == 3 and csll.to_list() == [1, 2, 3]

    # prepend
    csll.prepend(0)  # [0,1,2,3]
    show_csll("After prepend(0)", csll)
    assert csll.to_list() == [0, 1, 2, 3]

    # append
    csll.append(4)   # [0,1,2,3,4]
    show_csll("After append(4)", csll)
    assert csll.to_list() == [0, 1, 2, 3, 4]

    # insert_at middle
    csll.insert_at(3, 99)  # [0,1,2,99,3,4]
    show_csll("After insert_at(3,99)", csll)
    assert csll.to_list() == [0, 1, 2, 99, 3, 4]

    # insert_at head (0)
    csll.insert_at(0, -1)  # [-1,0,1,2,99,3,4]
    show_csll("After insert_at(0,-1)", csll)
    assert csll.to_list() == [-1, 0, 1, 2, 99, 3, 4]

    # insert_at end (len)
    csll.insert_at(len(csll), 5)  # [-1,0,1,2,99,3,4,5]
    show_csll("After insert_at(len,5)", csll)
    assert csll.to_list() == [-1, 0, 1, 2, 99, 3, 4, 5]

    # find existing & non-existing
    idx_99 = csll.find(99)
    idx_100 = csll.find(100)
    print(f"find(99) -> {idx_99}, find(100) -> {idx_100}")
    assert idx_99 == 4 and idx_100 == -1  # note index after earlier ops

    # get / set
    v4 = csll.get(4)  # should be 99
    print(f"get(4) -> {v4}")
    csll.set(4, 42)   # replace 99 with 42
    show_csll("After set(4,42)", csll)
    assert csll.to_list() == [-1, 0, 1, 2, 42, 3, 4, 5]

    # delete_at head
    removed = csll.delete_at(0)  # removes -1
    print(f"delete_at(0) -> {removed}")
    show_csll("After delete_at(0)", csll)
    assert removed == -1 and csll.to_list() == [0, 1, 2, 42, 3, 4, 5]

    # delete_at middle
    removed = csll.delete_at(3)  # removes 42
    print(f"delete_at(3) -> {removed}")
    show_csll("After delete_at(3)", csll)
    assert removed == 42 and csll.to_list() == [0, 1, 2, 3, 4, 5]

    # delete_at tail
    removed = csll.delete_at(len(csll) - 1)  # removes 5
    print(f"delete_at(len-1) -> {removed}")
    show_csll("After delete_at(len-1)", csll)
    assert removed == 5 and csll.to_list() == [0, 1, 2, 3, 4]

    # delete_value existing
    ok = csll.delete_value(2)
    print(f"delete_value(2) -> {ok}")
    show_csll("After delete_value(2)", csll)
    assert ok is True and csll.to_list() == [0, 1, 3, 4]

    # delete_value missing
    ok = csll.delete_value(999)
    print(f"delete_value(999) -> {ok}")
    show_csll("Attempt delete_value(999)", csll)
    assert ok is False and csll.to_list() == [0, 1, 3, 4]

    # clear & error cases
    csll.clear()
    show_csll("After clear()", csll)
    assert csll.is_empty() and csll.to_list() == []
    try:
        csll.delete_at(0)
    except IndexError as e:
        print(f"delete_at(0) on empty -> IndexError: {e}")
    try:
        csll.get(0)
    except IndexError as e:
        print(f"get(0) on empty -> IndexError: {e}")

    # quick rebuild
    csll.append(10); show_csll("append(10)", csll)
    csll.prepend(9); show_csll("prepend(9)", csll)
    csll.append(11); show_csll("append(11)", csll)
    assert csll.to_list() == [9, 10, 11]

    print("\n✅ All CSLL tests passed!")
````

## Circular Doubly Linked List

A **Circular Doubly Linked List (CDLL)** is a doubly linked list where the tail's `next` points to the head and the head's `prev` points to the tail. We often use a **sentinel** (dummy) node to simplify edge cases: the list is empty when `sentinel.next == sentinel`.

### Node Structure (Sentinel)
```
class Node {
  value       // data stored (sentinel has no data)
  Node prev   // pointer to previous node
  Node next   // pointer to next node
}
```
---
### Key Characteristics
- ✅ **Circular & bidirectional** for efficient deque-like ops.
- ✅ **Sentinel simplifies** insert/delete logic (no null checks at ends).
- ❌ More pointers per node; one extra sentinel node.


### Algorithms for Common Operations 🧑‍💻

Assume `S` is the sentinel; real head is `S.next` and real tail is `S.prev`.

#### Traversal — O(n)
```
function traverse(S):
  cur = S.next
  repeat n times:
    print(cur.value)
    cur = cur.next
```
#### Prepend — O(1)
```
function prepend(S, value):
  new = Node(value)
  new.prev = S
  new.next = S.next
  S.next.prev = new
  S.next = new
```
#### Append — O(1)
```
function append(S, value):
  new = Node(value)
  new.prev = S.prev
  new.next = S
  S.prev.next = new
  S.prev = new
```
| Operation | Time | Space | Notes |
|---|---|---|---|
| `is_empty` | O(1) | O(1) | sentinel check |
| `__len__` | O(1) | O(1) | size counter |
| `clear` | O(1) | O(1) | reset links to sentinel |
| `prepend` | O(1) | O(1) | after sentinel |
| `append` | O(1) | O(1) | before sentinel |
| `insert_at(i)` | O(n) | O(1) | walk from nearer side |
| `delete_at(i)` | O(n) | O(1) | unlink node |
| `delete_value(v)` | O(n) | O(1) | scan until match |
| `find(v)` | O(n) | O(1) | linear search |
| `get/set(i)` | O(n) | O(1) | walk to i |
| `to_list()` | O(n) | O(1) | gather values |


````python
# =====================================
# Circular Doubly Linked List (CDLL)
# =====================================

class CDLLNode:
    __slots__ = ("data","prev","next")                           # Limit attributes for memory efficiency
    def __init__(self, data=None):                               # Initialize a node; sentinel has data=None
        self.data = data                                         # Payload (None for sentinel)
        self.prev = self.next = self                             # Self-links create a 1-node circular list

class CircularDoublyLinkedList:
    def __init__(self, iterable=None):                           # Optional bulk construction
        self.sentinel = CDLLNode()                               # Create a sentinel that marks list boundaries
        self._size = 0                                           # Maintain element count (excludes sentinel)
        if iterable is not None:                                 # If initial values given
            for x in iterable:                                   # Iterate them
                self.append(x)                                   # Append each in O(1)

    def __len__(self):                                           # Length in O(1)
        return self._size                                        # Expose size counter

    def is_empty(self):                                          # Emptiness test
        return self._size == 0                                   # True if list holds zero real nodes

    def __repr__(self):                                          # Developer-friendly representation
        return f"CircularDoublyLinkedList({self.to_list()})"     # Use list form for clarity

    def clear(self):                                             # Remove all elements
        self.sentinel.next = self.sentinel                       # Point sentinel.next to itself
        self.sentinel.prev = self.sentinel                       # Point sentinel.prev to itself
        self._size = 0                                           # Reset size

    def _insert_between(self, value, prev_node, next_node):      # Core splice helper
        node = CDLLNode(value)                                   # Create real node with payload
        node.prev, node.next = prev_node, next_node              # Link node between prev_node and next_node
        prev_node.next = node                                    # Update previous neighbor's next
        next_node.prev = node                                    # Update next neighbor's prev
        self._size += 1                                          # Increase size counter

    def append(self, value):                                     # O(1) add before sentinel (tail position)
        self._insert_between(value, self.sentinel.prev, self.sentinel)  # Connect between old tail and sentinel

    def prepend(self, value):                                    # O(1) add after sentinel (head position)
        self._insert_between(value, self.sentinel, self.sentinel.next)  # Connect right after sentinel

    def insert_at(self, index, value):                           # Insert at index (0..size)
        if index < 0 or index > self._size:                      # Validate index
            raise IndexError("index out of range")               # Error if invalid
        if index <= self._size // 2:                             # Choose closer direction: head side
            cur = self.sentinel.next                             # Start at first real node
            for _ in range(index):                               # Walk forward index steps
                cur = cur.next                                   # Advance pointer
        else:                                                    # Closer from tail side
            cur = self.sentinel                                  # Start from sentinel
            for _ in range(self._size - index):                  # Walk backward appropriate steps
                cur = cur.prev                                   # Move pointer backward
        self._insert_between(value, cur.prev, cur)               # Splice new node before cur

    def delete_at(self, index):                                  # Delete node at index, return its value
        if self.is_empty():                                      # Empty => cannot delete
            raise IndexError("delete from empty list")           # Error
        if index < 0 or index >= self._size:                     # Validate index
            raise IndexError("index out of range")               # Error
        if index <= self._size // 2:                             # Walk from head if closer
            cur = self.sentinel.next                             # First real node
            for _ in range(index):                               # Move to index
                cur = cur.next                                   # Step forward
        else:                                                    # Walk from tail if closer
            cur = self.sentinel.prev                             # Last real node
            for _ in range(self._size - 1 - index):              # Move backward
                cur = cur.prev                                   # Step backward
        val = cur.data                                           # Save value to return
        cur.prev.next = cur.next                                 # Bridge previous to next
        cur.next.prev = cur.prev                                 # Bridge next to previous
        self._size -= 1                                          # Decrease size
        return val                                               # Return removed value

    def delete_value(self, value):                               # Remove first node equal to value
        cur = self.sentinel.next                                 # Start at head
        for _ in range(self._size):                              # Iterate exactly n nodes
            if cur.data == value:                                # Found a match
                cur.prev.next = cur.next                         # Unlink cur from list
                cur.next.prev = cur.prev                         # Mirror link for doubly list
                self._size -= 1                                  # Decrement size
                return True                                      # Deletion successful
            cur = cur.next                                       # Otherwise continue
        return False                                             # Value not found

    def find(self, value):                                       # Return index of first match or -1
        cur = self.sentinel.next                                 # Start at head
        for i in range(self._size):                              # Visit n nodes
            if cur.data == value:                                # Match?
                return i                                         # Return its index
            cur = cur.next                                       # Step forward
        return -1                                                # Not found

    def get(self, index):                                        # Random access by walking (O(n))
        if index < 0 or index >= self._size:                     # Validate index
            raise IndexError("index out of range")               # Error
        if index <= self._size // 2:                             # Choose side to start
            cur = self.sentinel.next                             # Head side
            for _ in range(index): cur = cur.next                # Move forward index steps
        else:                                                    # Tail side
            cur = self.sentinel.prev                             # Tail node
            for _ in range(self._size - 1 - index): cur = cur.prev  # Move backward
        return cur.data                                          # Return payload

    def set(self, index, value):                                 # Overwrite by index (O(n))
        if index < 0 or index >= self._size:                     # Validate
            raise IndexError("index out of range")               # Error
        if index <= self._size // 2:                             # Start closer to head
            cur = self.sentinel.next                             # Head node
            for _ in range(index): cur = cur.next                # Walk forward
        else:                                                    # Start closer to tail
            cur = self.sentinel.prev                             # Tail node
            for _ in range(self._size - 1 - index): cur = cur.prev  # Walk backward
        cur.data = value                                         # Assign payload

    def to_list(self):                                           # Convert to Python list
        out = []                                                 # Prepare container
        cur = self.sentinel.next                                 # Start at first real node
        for _ in range(self._size):                              # Loop n times
            out.append(cur.data)                                 # Collect payload
            cur = cur.next                                       # Step forward
        return out                                               # Return list

# ---------- Pretty printer for CDLL (bidirectional circular view) ----------
def as_circular_double_arrow(cdll):
    """
    Render as: [H]1 <-> 2 <-> 3 <-> (back to [H]); 'Empty' if no nodes.
    Head is the first real node: cdll.sentinel.next
    """
    if cdll.is_empty():
        return "Empty"
    parts = []
    cur = cdll.sentinel.next  # first real node
    n = len(cdll)
    for i in range(n):
        parts.append(f"[H]{cur.data}" if i == 0 else str(cur.data))
        cur = cur.next
    return " <-> ".join(parts) + " <-> (back to [H])"

def show_cdll(stage, cdll):
    """Print stage label + current CDLL + size."""
    print(f"{stage:<34} | {as_circular_double_arrow(cdll)}  (len={len(cdll)})")

# ---------- Test Suite for CircularDoublyLinkedList ----------
def run_cdll_tests():
    print("=== CircularDoublyLinkedList Tests ===")

    # Init from iterable
    cdll = CircularDoublyLinkedList([1, 2, 3])
    show_cdll("Init from [1,2,3]", cdll)
    assert len(cdll) == 3 and cdll.to_list() == [1, 2, 3]

    # prepend
    cdll.prepend(0)  # [0,1,2,3]
    show_cdll("After prepend(0)", cdll)
    assert cdll.to_list() == [0, 1, 2, 3]

    # append
    cdll.append(4)   # [0,1,2,3,4]
    show_cdll("After append(4)", cdll)
    assert cdll.to_list() == [0, 1, 2, 3, 4]

    # insert_at middle
    cdll.insert_at(3, 99)  # [0,1,2,99,3,4]
    show_cdll("After insert_at(3,99)", cdll)
    assert cdll.to_list() == [0, 1, 2, 99, 3, 4]

    # insert_at head (0)
    cdll.insert_at(0, -1)  # [-1,0,1,2,99,3,4]
    show_cdll("After insert_at(0,-1)", cdll)
    assert cdll.to_list() == [-1, 0, 1, 2, 99, 3, 4]

    # insert_at end (len)
    cdll.insert_at(len(cdll), 5)  # [-1,0,1,2,99,3,4,5]
    show_cdll("After insert_at(len,5)", cdll)
    assert cdll.to_list() == [-1, 0, 1, 2, 99, 3, 4, 5]

    # find existing & non-existing
    idx_99 = cdll.find(99)
    idx_100 = cdll.find(100)
    print(f"find(99) -> {idx_99}, find(100) -> {idx_100}")
    assert idx_99 == 4 and idx_100 == -1

    # get / set
    v4 = cdll.get(4)  # currently 99
    print(f"get(4) -> {v4}")
    cdll.set(4, 42)   # replace 99 with 42
    show_cdll("After set(4,42)", cdll)
    assert cdll.to_list() == [-1, 0, 1, 2, 42, 3, 4, 5]

    # delete_at head
    removed = cdll.delete_at(0)  # remove -1
    print(f"delete_at(0) -> {removed}")
    show_cdll("After delete_at(0)", cdll)
    assert removed == -1 and cdll.to_list() == [0, 1, 2, 42, 3, 4, 5]

    # delete_at middle
    removed = cdll.delete_at(3)  # remove 42
    print(f"delete_at(3) -> {removed}")
    show_cdll("After delete_at(3)", cdll)
    assert removed == 42 and cdll.to_list() == [0, 1, 2, 3, 4, 5]

    # delete_at tail
    removed = cdll.delete_at(len(cdll) - 1)  # remove 5
    print(f"delete_at(len-1) -> {removed}")
    show_cdll("After delete_at(len-1)", cdll)
    assert removed == 5 and cdll.to_list() == [0, 1, 2, 3, 4]

    # delete_value existing
    ok = cdll.delete_value(2)
    print(f"delete_value(2) -> {ok}")
    show_cdll("After delete_value(2)", cdll)
    assert ok is True and cdll.to_list() == [0, 1, 3, 4]

    # delete_value missing
    ok = cdll.delete_value(999)
    print(f"delete_value(999) -> {ok}")
    show_cdll("Attempt delete_value(999)", cdll)
    assert ok is False and cdll.to_list() == [0, 1, 3, 4]

    # clear & error cases
    cdll.clear()
    show_cdll("After clear()", cdll)
    assert cdll.is_empty() and cdll.to_list() == []
    try:
        cdll.delete_at(0)
    except IndexError as e:
        print(f"delete_at(0) on empty -> IndexError: {e}")
    try:
        cdll.get(0)
    except IndexError as e:
        print(f"get(0) on empty -> IndexError: {e}")

    # quick rebuild
    cdll.append(10); show_cdll("append(10)", cdll)
    cdll.prepend(9); show_cdll("prepend(9)", cdll)
    cdll.append(11); show_cdll("append(11)", cdll)
    assert cdll.to_list() == [9, 10, 11]

    print("\n✅ All CDLL tests passed!")
````

## Stack

A **Stack** is a **LIFO (Last-In, First-Out)** ADT supporting operations on the top of the collection.  
We implement both an **array-backed** version (dynamic array) and a **linked-list-backed** version.

### Key Characteristics
- ✅ `push`/`pop` on the top only.
- ✅ `peek` reads the top without removal.
- ❌ No random insertions (only at top).

### Algorithms for Common Operations 🧑‍💻

#### Array-backed (dynamic array)
- `push(x)` / `pop()` are **amortized O(1)** due to occasional resizing.
```
function push(A, x): append x to A
function pop(A): if A empty -> error; else return last element
```
#### Linked-list-backed
- Always **O(1)** since we modify the head pointer.
```
function push(head, x): new = Node(x); new.next = head; head = new
function pop(head):  if head == null -> error; x = head.value; head = head.next
```

| Operation | Time | Space | Notes |
|---|---|---|---|
| `push(x)` | O(1)* | O(1) | Array amortized; Linked exact |
| `pop()` | O(1)* | O(1) | |
| `peek()` | O(1) | O(1) | |
| `is_empty` / `size` / `clear` | O(1) | O(1) | |


````python
# ==================
# Stack Implementations
# ==================

class ArrayStack:
    def __init__(self, iterable=None):                          # Optional bulk construction
        self._a = []                                            # Underlying dynamic Python list
        if iterable is not None:                                # If initial values given
            for x in iterable:                                  # Iterate through them
                self.push(x)                                    # Use push for uniform behavior

    def push(self, x):                                          # Push to top in amortized O(1)
        self._a.append(x)                                       # Python list append is amortized O(1)

    def pop(self):                                              # Pop from top in amortized O(1)
        if not self._a:                                         # If stack is empty
            raise IndexError("pop from empty stack")            # Raise underflow error
        return self._a.pop()                                    # Remove and return last element

    def peek(self):                                             # Read top without removing
        if not self._a:                                         # Empty check
            raise IndexError("peek from empty stack")           # Error if empty
        return self._a[-1]                                      # Last element is the top

    def is_empty(self):                                         # Emptiness test
        return len(self._a) == 0                                # True if array length is zero

    def size(self):                                             # Size of stack
        return len(self._a)                                     # Return array length

    def clear(self):                                            # Remove all elements
        self._a.clear()                                         # Python list clear is O(1)

    def __repr__(self):                                         # Developer-friendly representation
        return f"ArrayStack({self._a})"                         # Display underlying list

class StackNode:
    __slots__ = ("data","next")                                # Node fields for linked stack
    def __init__(self, data, next=None):                         # Init a stack node
        self.data = data                                        # Payload
        self.next = next                                        # Next pointer (towards old head)

class LinkedStack:
    def __init__(self, iterable=None):                           # Optional bulk construction
        self.head = None                                        # Top of stack (None if empty)
        self._size = 0                                          # Size counter
        if iterable is not None:                                # Preload if provided
            for x in iterable:                                  # Iterate values
                self.push(x)                                    # Push each value

    def push(self, x):                                          # O(1) push at head
        self.head = StackNode(x, self.head)                     # Allocate node and link to old head
        self._size += 1                                         # Increase size

    def pop(self):                                              # O(1) pop at head
        if not self.head:                                       # If stack empty
            raise IndexError("pop from empty stack")            # Underflow
        x = self.head.data                                      # Capture top value
        self.head = self.head.next                              # Move head to next node
        self._size -= 1                                         # Decrease size
        return x                                                # Return popped value

    def peek(self):                                             # O(1) read top
        if not self.head:                                       # If empty
            raise IndexError("peek from empty stack")           # Error
        return self.head.data                                   # Return top without removing

    def is_empty(self):                                         # O(1) emptiness
        return self._size == 0                                  # True if size is zero

    def size(self):                                             # O(1) size
        return self._size                                       # Return maintained size

    def clear(self):                                            # O(1) clear
        self.head = None                                        # Drop head reference
        self._size = 0                                          # Reset size

    def __repr__(self):                                         # Debug representation
        vals = []                                               # Collect values top->down
        cur = self.head                                         # Start at head
        while cur is not None:                                  # Traverse nodes
            vals.append(cur.data)                               # Accumulate payload
            cur = cur.next                                      # Move to next
        return f"LinkedStack(top-> {vals})"                     # Show order from top

# ========== Vertical printer for Stack ==========
def _stack_items(stack):
    """Return items from TOP to BOTTOM as a Python list."""
    # Array-backed: top is at end of the list -> print top first
    if hasattr(stack, "_a"):
        return list(reversed(stack._a))
    # Linked-backed: head is the top -> traverse forward
    vals, cur = [], getattr(stack, "head", None)
    while cur is not None:
        vals.append(cur.data)
        cur = cur.next
    return vals

def print_stack_vertical(stack):
    """Print:
    TOP
    │ x │
    │ y │
    ...
    BOTTOM
    """
    items = _stack_items(stack)
    if not items:
        print("Stack is Empty\n")
        return
    print("TOP")
    for item in items:
        print(f"│ {item} │")
    print("BOTTOM\n")

def show_stack(stage, stack):
    """Stage label + vertical view + size/peek (if available)."""
    print(f"{stage}")
    print_stack_vertical(stack)
    print(f"(len={stack.size()})\n")

# ========== Tests for ArrayStack & LinkedStack ==========
def run_stack_tests():
    print("=== Stack Tests (ArrayStack) ===")
    a = ArrayStack([10, 20, 30])
    show_stack("Init ArrayStack [10,20,30]", a)
    a.push(40); show_stack("After push(40)", a)
    assert a.peek() == 40
    print(f"peek() -> {a.peek()}\n")
    assert a.pop() == 40
    show_stack("After pop()", a)
    a.clear(); show_stack("After clear()", a)
    try:
        a.pop()
    except IndexError as e:
        print(f"pop() on empty -> IndexError: {e}\n")

    print("=== Stack Tests (LinkedStack) ===")
    l = LinkedStack([1, 2, 3])
    show_stack("Init LinkedStack [1,2,3]", l)
    l.push(4); show_stack("After push(4)", l)
    assert l.peek() == 4
    print(f"peek() -> {l.peek()}\n")
    assert l.pop() == 4
    show_stack("After pop()", l)
    l.clear(); show_stack("After clear()", l)
    try:
        l.peek()
    except IndexError as e:
        print(f"peek() on empty -> IndexError: {e}\n")

    print("✅ Stack tests passed!\n")
````

````python
# Run
run_stack_tests()
````

## Queue

A **Queue** is a **FIFO (First-In, First-Out)** ADT where insertion happens at the **rear** and removal at the **front**. We implement an **array-backed circular buffer** and a **linked-list** queue.

### Key Characteristics
- ✅ Enqueue at tail, dequeue at head.
- ✅ `front()` and `rear()` read without removal.
- ❌ No random access; only ends are efficient.

### Algorithms for Common Operations 🧑‍💻

#### Array-backed (circular buffer) — amortized O(1)
```
enqueue: if full -> grow; A[tail]=x; tail=(tail+1)%cap; size++
dequeue: if empty -> error; x=A[head]; A[head]=null; head=(head+1)%cap; size--
```
#### Linked-list-backed — O(1)
```
enqueue: node=Node(x); tail.next=node (or head=node); tail=node
dequeue: if head null -> error; x=head.value; head=head.next; if head null -> tail=null
```

| Operation | Time | Space | Notes |
|---|---|---|---|
| `enqueue(x)` | O(1)* | O(1) | Array amortized; Linked exact |
| `dequeue()` | O(1)* | O(1) | |
| `front()` / `rear()` | O(1) | O(1) | |
| `is_empty` / `size` / `clear` | O(1) | O(1) | |


````python
# =====================
# Queue Implementations
# =====================

class ArrayQueue:
    def __init__(self, capacity=8):                             # Initialize with a starting capacity
        if capacity <= 0:                                       # Capacity must be positive
            raise ValueError("capacity must be positive")       # Validate capacity
        self._a = [None] * capacity                             # Fixed-size list holding queued elements
        self._head = 0                                          # Index of current front
        self._tail = 0                                          # Index of next insertion position
        self._size = 0                                          # Number of elements in queue

    def _grow(self):                                            # Double the buffer capacity
        new_cap = len(self._a) * 2                              # Compute new capacity
        new_a = [None] * new_cap                                # Allocate a larger list
        for i in range(self._size):                             # Copy current elements in order
            new_a[i] = self._a[(self._head + i) % len(self._a)] # Preserve logical order relative to head
        self._a = new_a                                         # Switch to the new buffer
        self._head = 0                                          # Reset head to index 0
        self._tail = self._size                                 # Tail becomes size (next insertion slot)

    def enqueue(self, x):                                       # Add to rear (amortized O(1))
        if self._size == len(self._a):                          # If buffer is full
            self._grow()                                        # Grow capacity
        self._a[self._tail] = x                                 # Place element at tail position
        self._tail = (self._tail + 1) % len(self._a)            # Advance tail circularly
        self._size += 1                                         # Increase element count

    def dequeue(self):                                          # Remove from front (amortized O(1))
        if self._size == 0:                                     # Underflow check
            raise IndexError("dequeue from empty queue")        # Error if empty
        x = self._a[self._head]                                 # Read front element
        self._a[self._head] = None                              # Help GC by clearing slot
        self._head = (self._head + 1) % len(self._a)            # Advance head circularly
        self._size -= 1                                         # Decrease count
        return x                                                # Return dequeued value

    def front(self):                                            # Peek front in O(1)
        if self._size == 0:                                     # Empty check
            raise IndexError("front from empty queue")          # Error
        return self._a[self._head]                              # Return current front element

    def rear(self):                                             # Peek rear in O(1)
        if self._size == 0:                                     # Empty check
            raise IndexError("rear from empty queue")           # Error
        return self._a[(self._tail - 1) % len(self._a)]         # Return element just before tail index

    def is_empty(self):                                         # O(1) emptiness
        return self._size == 0                                  # True if no elements

    def size(self):                                             # O(1) size
        return self._size                                       # Return count

    def clear(self):                                            # O(1) clear (reset state)
        self._a = [None] * 8                                    # Reinitialize buffer to default capacity
        self._head = self._tail = self._size = 0                # Reset indices and size

    def __repr__(self):                                         # Developer-friendly representation
        if self._size == 0:                                     # If queue empty
            return "ArrayQueue([])"                             # Show empty
        data = [self._a[(self._head+i)%len(self._a)] for i in range(self._size)]  # Rebuild logical order
        return f"ArrayQueue({data})"                            # Print values in order

class QNode:
    __slots__ = ("data","next")                                # Node fields for linked queue
    def __init__(self, data, next=None):                         # Initialize node
        self.data = data                                        # Payload
        self.next = next                                        # Pointer to next node

class LinkedQueue:
    def __init__(self, iterable=None):                           # Optional bulk construction
        self.head = self.tail = None                            # Head points to front, tail to rear
        self._size = 0                                          # Count of elements
        if iterable is not None:                                # Preload
            for x in iterable:                                  # Iterate initial values
                self.enqueue(x)                                 # Enqueue each

    def enqueue(self, x):                                       # O(1) add to rear
        node = QNode(x)                                         # Create a node
        if self.tail:                                           # If non-empty
            self.tail.next = node                               # Link old tail to new node
        else:                                                   # If empty
            self.head = node                                    # New node is also the head
        self.tail = node                                        # Update tail to new node
        self._size += 1                                         # Increase size

    def dequeue(self):                                          # O(1) remove from front
        if not self.head:                                       # If queue empty
            raise IndexError("dequeue from empty queue")        # Underflow
        x = self.head.data                                      # Read front value
        self.head = self.head.next                              # Move head to next node
        if not self.head:                                       # If we just removed last element
            self.tail = None                                    # Tail should also become None
        self._size -= 1                                         # Decrease size
        return x                                                # Return dequeued value

    def front(self):                                            # O(1) peek front
        if not self.head:                                       # Empty check
            raise IndexError("front from empty queue")          # Error
        return self.head.data                                   # Value at head

    def rear(self):                                             # O(1) peek rear
        if not self.tail:                                       # Empty check
            raise IndexError("rear from empty queue")           # Error
        return self.tail.data                                   # Value at tail

    def is_empty(self):                                         # O(1) emptiness
        return self._size == 0                                  # True if count is zero

    def size(self):                                             # O(1) size
        return self._size                                       # Return stored count

    def clear(self):                                            # O(1) clear
        self.head = self.tail = None                            # Drop references to nodes
        self._size = 0                                          # Reset size

    def __repr__(self):                                         # Debug-friendly representation
        vals = []                                               # Collect payloads
        cur = self.head                                         # Start at front
        while cur is not None:                                  # Traverse nodes
            vals.append(cur.data)                               # Append payload
            cur = cur.next                                      # Move forward
        return f"LinkedQueue({vals})"                           # Show in list form

# ========== Vertical printer for Queue ==========
def _queue_items(q):
    """Return items from FRONT to REAR as a Python list."""
    # LinkedQueue: traverse from head to tail
    if hasattr(q, "head"):
        vals, cur = [], q.head
        while cur is not None:
            vals.append(cur.data)
            cur = cur.next
        return vals
    # ArrayQueue: respect circular order from head to head+size-1
    vals = []
    cap = len(q._a)
    for i in range(q._size):
        vals.append(q._a[(q._head + i) % cap])
    return vals

def print_queue_vertical(q):
    """Print:
    FRONT
    │ x │
    │ y │
    ...
    REAR
    """
    items = _queue_items(q)
    if not items:
        print("Queue is Empty\n")
        return
    print("FRONT")
    for item in items:
        print(f"│ {item} │")
    print("REAR\n")

def show_queue(stage, q):
    """Stage label + vertical view + size/front/rear (if available)."""
    print(stage)
    print_queue_vertical(q)
    # front()/rear() raise on empty — guard them:
    f = q.front() if not q.is_empty() else None
    r = q.rear()  if not q.is_empty() else None
    print(f"(len={q.size()}, front={f}, rear={r})\n")

# ========== Tests for ArrayQueue & LinkedQueue ==========
def run_queue_tests():
    print("=== Queue Tests (ArrayQueue) ===")
    aq = ArrayQueue(capacity=2)     # small cap to exercise _grow
    show_queue("Init ArrayQueue (cap=2)", aq)
    aq.enqueue(10); show_queue("After enqueue(10)", aq)
    aq.enqueue(20); show_queue("After enqueue(20)", aq)
    aq.enqueue(30); show_queue("After enqueue(30)  # triggers grow", aq)
    assert aq.front() == 10 and aq.rear() == 30
    print(f"dequeue() -> {aq.dequeue()}")  # 10
    show_queue("After dequeue()", aq)
    assert aq.front() == 20
    aq.clear(); show_queue("After clear()", aq)
    try:
        aq.dequeue()
    except IndexError as e:
        print(f"dequeue() on empty -> IndexError: {e}\n")

    print("=== Queue Tests (LinkedQueue) ===")
    lq = LinkedQueue([1, 2, 3])
    show_queue("Init LinkedQueue [1,2,3]", lq)
    lq.enqueue(4); show_queue("After enqueue(4)", lq)
    assert lq.front() == 1 and lq.rear() == 4
    print(f"dequeue() -> {lq.dequeue()}")  # 1
    show_queue("After dequeue()", lq)
    lq.clear(); show_queue("After clear()", lq)
    try:
        lq.front()
    except IndexError as e:
        print(f"front() on empty -> IndexError: {e}\n")

    print("✅ Queue tests passed!\n")
````

## Trees
A Tree is a hierarchical, non-linear data structure made of nodes connected by edges. One node is the root; every other node has exactly one parent and zero or more children. Trees model “part-of” relationships (filesystems, DOM, org charts, parsers, indices).

Types of Trees:
- Binary Tree
- Binary Search Tree
- AVL Tree
- Red-Black Tree
- B-Tree
- Trie

Types of binary trees:
- **Full Binary Tree:** Each node has 0 or 2 children.
- **Complete Binary Tree:** All levels filled except possibly last, filled left to right.
- **Perfect Binary Tree:** All internal nodes have 2 children and all leaves are at the same level.
- **Balanced Binary Tree:** Height difference between left and right subtrees is at most 1 for every node.

````python
class TreeNode:
    def __init__(self, data, children=[]):
        self.data = data
        self.children = children if children is not None else []

    def __str__(self, level=0):
        ret = "  " * level + repr(self.data) + "\n"
        for child in self.children:
            ret += child.__str__(level + 1)
        return ret
    
    def addChild(self, child_node):
        self.children.append(child_node)

tree = TreeNode("Drinks",[])
cold = TreeNode("Cold",[])
hot = TreeNode("Hot",[])
tree.addChild(cold)
tree.addChild(hot)
cold.addChild(TreeNode("Coke",[]))
cold.addChild(TreeNode("Pepsi",[]))
cold.addChild(TreeNode("Fanta",[]))
hot.addChild(TreeNode("Tea",[]))
hot.addChild(TreeNode("Coffee",[]))
print(tree)

````

````python
class BinaryTreeLLNode:
    def __init__(self,data):
        self.data = data
        self.left = None
        self.right = None
    
    def __str__(self, level=0):
        ret = "  " * level + repr(self.data) + "\n"
        if self.left:
            ret += self.left.__str__(level + 1)
        if self.right:
            ret += self.right.__str__(level + 1)
        return ret

    def preorder(self):
        print(self.data)
        if self.left:
            self.left.preorder()
        if self.right:
            self.right.preorder()
    
    def inorder(self):
        if self.left:
            self.left.inorder()
        print(self.data)
        if self.right:
            self.right.inorder()
    
    def postorder(self):
        if self.left:
            self.left.postorder()
        if self.right:
            self.right.postorder()
        print(self.data)
    
    def level_order(self):
        queue = []
        queue.append(self)
        while queue:
            current = queue.pop(0)
            print(current.data)
            if current.left:
                queue.append(current.left)
            if current.right:
                queue.append(current.right)
    
    def search(self, value):
        if self.data == value:
            return True
        left_found = self.left.search(value) if self.left else False
        right_found = self.right.search(value) if self.right else False
        return left_found or right_found

    def insert(self,value):
        queue = []
        queue.append(self)
        while queue:
            current = queue.pop(0)
            if not current.left:
                current.left = BinaryTreeLLNode(value)
                return
            else:
                queue.append(current.left)
            if not current.right:
                current.right = BinaryTreeLLNode(value)
                return
            else:
                queue.append(current.right)

newTree = BinaryTreeLLNode("Drinks")
cold = BinaryTreeLLNode("Cold")
hot = BinaryTreeLLNode("Hot")
newTree.left = cold
newTree.right = hot
cold.left = BinaryTreeLLNode("Coke")
cold.right = BinaryTreeLLNode("Pepsi")
cold.right = BinaryTreeLLNode("Fanta")
hot.left = BinaryTreeLLNode("Tea")
hot.right = BinaryTreeLLNode("Coffee")
print("Preorder Traversal:")
newTree.preorder()
print("\nInorder Traversal:")
newTree.inorder()
print("\nPostorder Traversal:")
newTree.postorder()
print("\nLevel-order Traversal:")
newTree.level_order()

print("\nSearch for 'Tea':", newTree.search("Tea"))

print("\nInserting 'Juice' into the tree.")
newTree.insert("Juice")
print("\nLevel-order Traversal after insertion:")
newTree.level_order()

print(newTree)
````

## Binary Search Tree (BST)
A **Binary Search Tree (BST)** is a binary tree where each node follows the property:
- Left subtree contains only nodes with values less than the node's value.
- Right subtree contains only nodes with values greater than the node's value.

### Time and Space Complexity of Binary Search Tree (BST) Operations

| Operation      | Average Case | Worst Case      | Space Complexity |
|----------------|-------------|-----------------|------------------|
| Search         | O(log n)    | O(n)            | O(1)             |
| Insertion      | O(log n)    | O(n)            | O(1)             |
| Deletion       | O(log n)    | O(n)            | O(1)             |
| Traversal      | O(n)        | O(n)            | O(n) (recursion) |

- *Average case assumes a balanced BST. Worst case occurs when the tree is skewed (like a linked list).*


````python
class BSTNode:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

    def insert(self,value):
        if self.data is None:
            self.data = value
        
        if value < self.data:
            if self.left is None:
                self.left = BSTNode(value)
            else:
                self.left.insert(value)
        else:
            if self.right is None:
                self.right = BSTNode(value)
            else:
                self.right.insert(value)
        return f"Inserted {value}"
    
    def preOrderTraversal(self):
        print(self.data,end=" ")
        if self.left:
            self.left.preOrderTraversal()
        if self.right:
            self.right.preOrderTraversal()

    def inOrderTraversal(self):
        if self.left:
            self.left.inOrderTraversal()
        print(self.data,end=" ")
        if self.right:
            self.right.inOrderTraversal()
    
    def postOrderTraversal(self):
        if self.left:
            self.left.postOrderTraversal()
        if self.right:
            self.right.postOrderTraversal()
        print(self.data,end=" ")
    
    def levelOrderTraversal(self):
        queue = []
        queue.append(self)
        while queue:
            current = queue.pop(0)
            print(current.data,end=" ")
            if current.left:
                queue.append(current.left)
            if current.right:
                queue.append(current.right)
    def search(self,value):
        if self.data == value:
            return f"Found {value}"
        elif value < self.data:
            if self.left:  # Check if the left child exists
                return self.left.search(value)  # Return the result of the recursive search
            else:
                return "Not Found"
        else:
            if self.right:
                return self.right.search(value)
            else:
                return "Not Found"
    
    def delete(self,value):
        if value < self.data:
            if self.left:
                self.left = self.left.delete(value)
        elif value > self.data:
            if self.right:
                self.right = self.right.delete(value)
        else: # when it is equal then delete it 
            if self.left is None:
                return self.right
            elif self.right is None:
                return self.left
            # find the smallest in the right sub tree and make that as the root 
            temp = self.right
            while temp.left:
                temp = temp.left
            self.data = temp.data
            self.right = self.right.delete(temp.data)
        return self
        
    
treeBst = BSTNode(40)  # Root node
treeBst.insert(20)     # Insert into the left subtree
treeBst.insert(60)     # Insert into the right subtree
treeBst.insert(10)     # Insert into the left subtree of 20
treeBst.insert(30)     # Insert into the right subtree of 20
treeBst.insert(50)     # Insert into the left subtree of 60
treeBst.insert(70)     # Insert into the right subtree of 60
treeBst.insert(15)
treeBst.insert(5)
print("\nPre-Order: ")
treeBst.preOrderTraversal()
print("\nIn-Order: ")
treeBst.inOrderTraversal()
print("\nPost-Order: ")
treeBst.postOrderTraversal()
print("\nLevel Order Traversal")
treeBst.levelOrderTraversal()

print("\n\n"+treeBst.search(150))
print(treeBst.search(40))
print(treeBst.search(10))
````

## AVL Tree
An **AVL Tree** is a self-balancing binary search tree where the heights of the two child subtrees of any node differ by at most one. If the balance factor (height of left subtree - height of right subtree) becomes greater than 1 or less than -1, the tree is rebalanced using rotations.

Insertion and deletion operations may require one or more rotations to maintain balance.
-  LL Rotation: Single right rotation.
-  RR Rotation: Single left rotation.
-  LR Rotation: Left rotation followed by right rotation.
-  RL Rotation: Right rotation followed by left rotation.
### AVL Tree — LL Rotation (2 Simple Medium-Tree Insertion Examples) 😊

#### Example 1 — **Internal (subtree) LL rotation**

* **Goal**: Show LL rotation at an internal node (not the root).
* **Insert sequence**: Start with a medium balanced-ish tree, then insert **10**.

**Initial tree**

```
            50
          /     \
        30       70
       /        /  \
     20        60  80
```

**After inserting 10 (before rebalancing)**

```
            50
          /     \
        30       70
       /        /  \
     20        60  80
    /
  10

First unbalanced node on the path up from 10: y = 30
Pattern: inserted into y.left.left  =>  LL case at node 30
Action: single RIGHT rotation at 30
```

**After LL rotation at node 30**

```
            50
          /     \
        20       70
       /  \     /  \
     10   30   60  80
```

* **Effect**: Subtree rooted at 30 rebalances; BST order preserved.

---

#### Example 2 — **Root LL rotation**

* **Goal**: Show LL rotation when the **root** becomes unbalanced.
* **Insert sequence**: Insert **10** into the left subtree to tip the root.

**Initial tree**

```
            50
          /     \
        30       65
       /  \
     20   40
```

**After inserting 10 (before rebalancing)**

```
            50
          /     \
        30       65
       /  \
     20   40
    /
  10

First unbalanced node on the path up from 10: y = 50 (the root)
Pattern: inserted into y.left.left  =>  LL case at ROOT
Action: single RIGHT rotation at 50
```

**After LL rotation at the root (50)**

```
            30
           /  \
         20    50
        /     /  \
      10     40  65
```

* **Effect**: The **root changes** to 30; the whole tree is balanced again.

---

#### Quick tip

* LL trigger: the first unbalanced node `y` has `bf(y)=+2` *and* the insertion was in `y.left.left`.
* Fix: **Right-rotate at `y`**, update heights, and you’re done for a single insertion path.


````python
class AVLTree:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None
        self.height = 1  # New node is initially added at leaf

    def preOrderTraversal(self):
        print(self.data,end=" ")
        if self.left:
            self.left.preOrderTraversal()
        if self.right:
            self.right.preOrderTraversal()

    def inOrderTraversal(self):
        if self.left:
            self.left.inOrderTraversal()
        print(self.data,end=" ")
        if self.right:
            self.right.inOrderTraversal()

    def postOrderTraversal(self):
        if self.left:
            self.left.postOrderTraversal()
        if self.right:
            self.right.postOrderTraversal()
        print(self.data,end=" ")

    def levelOrderTraversal(self):
        queue = []
        queue.append(self)
        while queue:
            current = queue.pop(0)
            print(current.data,end=" ")
            if current.left:
                queue.append(current.left)
            if current.right:
                queue.append(current.right)

    def search(self,value):
        if self.data == value:
            return f"Found {value}"
        elif value < self.data:
            if self.left:  # Check if the left child exists
                return self.left.search(value)  # Return the result of the recursive search
            else:
                return "Not Found"
        else:
            if self.right:
                return self.right.search(value)
            else:
                return "Not Found"
    def getHeight(self,node):
        if not node:
            return 0
        return node.height

    def rightRotate(self):
        newroot = self.left
        self.left = self.left.right
        newroot.right = self
        self.height = 1 + max(self.getHeight(self.left),self.getHeight(self.right)) # we are adding 1 here because height of node is 1 + max height of left and right subtrees
        newroot.height = 1 + max(self.getHeight(newroot.left),self.getHeight(newroot.right))
        return newroot
    
    def leftRotate(self):
        newroot = self.right
        self.right = self.right.left
        newroot.left = self
        self.height = 1 + max(self.getHeight(self.left),self.getHeight(self.right)) # we are adding 1 here because height of node is 1 + max height of left and right subtrees
        newroot.height = 1 + max(self.getHeight(newroot.left),self.getHeight(newroot.right))
        return newroot
    
    def getBalance(self):
        if not self:
            return 0
        return self.getHeight(self.left) - self.getHeight(self.right)
    
    def insert(self,value):
        if self.data is None:
            self.data = value
        
        if value < self.data:
            if self.left is None:
                self.left = AVLTree(value)
            else:
                self.left = self.left.insert(value)
        elif value > self.data:
            if self.right is None:
                self.right = AVLTree(value)
            else:
                self.right = self.right.insert(value)
        else:
            return self # Handle duplicate values by doing nothing
        
        self.height = 1 + max(self.getHeight(self.left),self.getHeight(self.right))
        balance = self.getBalance()

        # Left Left Case
        if balance > 1 and value < self.left.data:
            return self.rightRotate()
        
        # Right Right Case
        if balance < -1 and value > self.right.data:
            return self.leftRotate()
        
        # Left Right Case
        if balance > 1 and value > self.left.data:
            self.left = self.left.leftRotate()
            return self.rightRotate()
        
        # Right Left Case
        if balance < -1 and value < self.right.data:
            self.right = self.right.rightRotate()
            return self.leftRotate()
        return self

    def getMinValueNode(self,node):
        if node is None or node.left is None:
            return node
        return self.getMinValueNode(node.left)

    def delete(self,value):
        if not self:
            return self
        
        if value < self.data:
            if self.left:
                self.left = self.left.delete(value)
        elif value > self.data:
            if self.right:
                self.right = self.right.delete(value)
        else:
            if self.left is None:
                return self.right
            elif self.right is None:
                return self.left
            temp = self.getMinValueNode(self.right)
            self.data = temp.data
            self.right = self.right.delete(temp.data)
        
        self.height = 1 + max(self.getHeight(self.left),self.getHeight(self.right))
        balance = self.getBalance()

        # Left Left Case
        if balance > 1 and self.left.getBalance() >= 0:
            return self.rightRotate()
        
        # Right Right Case
        if balance < -1 and self.right.getBalance() <= 0:
            return self.leftRotate()
        
        # Left Right Case
        if balance > 1 and self.left.getBalance() < 0:
            self.left = self.left.leftRotate()
            return self.rightRotate()
        
        # Right Left Case
        if balance < -1 and self.right.getBalance() > 0:
            self.right = self.right.rightRotate()
            return self.leftRotate()
        
        return self
    
    def delteTree(self):
        self.data = None
        self.left = None
        self.right = None
        self.height = 0
        return "Tree Deleted"

avlTree = AVLTree(30)
avlTree = avlTree.insert(25)
avlTree = avlTree.insert(35)
avlTree = avlTree.insert(20)
avlTree = avlTree.insert(15)
avlTree = avlTree.insert(5)
avlTree.levelOrderTraversal()
print()
avlTree.preOrderTraversal()
print()
avlTree.inOrderTraversal()
````

## Binary Heap
A **Binary Heap** is a complete binary tree that satisfies the heap property:
- **Max-Heap**: Every parent node is greater than or equal to its children.
- **Min-Heap**: Every parent node is less than or equal to its children.
### Time and Space Complexity of Binary Heap Operations
| Operation      | Time Complexity | Space Complexity |
|----------------|-----------------|------------------|
| Insertion      | O(log n)       | O(1)             |
| Deletion (Max/Min) | O(log n)   | O(1)             |
| Peek (Max/Min) | O(1)           | O(1)             |  
| Build Heap     | O(n)           | O(1)             |
- *Binary heaps are typically implemented using arrays for efficient storage and access.*

````python
class Heap:
    def __init__(self,size,heapType="Min"):
        self.heap = (size+1)*[None]  # Initialize an empty list to store heap elements
        self.size = 0
        self.maxSize = size + 1
        self.type = heapType
    
    def peek(self):
        if not self.heap:
            return "Heap is empty"
        return self.heap[1]

    def sizeOfHeap(self):
        return self.size
    
    def levelOrderTraversal(self):
        for i in range(1,self.size+1):
            print(self.heap[i],end=" ")
        print()
    
    def heapifyTreeInsert(self,index):
        parentIndex = index // 2
        if index <= 1:
            return
        if self.type == "Min":
            if self.heap[index] < self.heap[parentIndex]:
                temp = self.heap[index]
                self.heap[index] = self.heap[parentIndex]
                self.heap[parentIndex] = temp
            self.heapifyTreeInsert(parentIndex)
        else:  # Max Heap
            if self.heap[index] > self.heap[parentIndex]:
                temp = self.heap[index]
                self.heap[index] = self.heap[parentIndex]
                self.heap[parentIndex] = temp
            self.heapifyTreeInsert(parentIndex)
    
    def insertNode(self,value):
        if self.size + 1 == self.maxSize:
            return "The heap is full"
        self.heap[self.size + 1] = value
        self.size += 1
        self.heapifyTreeInsert(self.size)
        return "Value inserted successfully"
    
    def heapifyTreeExtract(self,index):
        """
        Restore heap property after extraction by moving the node at 'index' down the tree.
        This is called "heapify down" or "sift down" operation.
        """
        leftIndex = index * 2      # Left child is at 2*parent_index
        rightIndex = index * 2 + 1 # Right child is at 2*parent_index + 1
        swapChild = 0              # Will store index of child to swap with
        
        # Case 1: Node has no children (it's a leaf node)
        if self.size < leftIndex:
            return  # Nothing to do, heap property is satisfied
        
        # Case 2: Node has only left child (no right child)
        elif self.size == leftIndex:
            if self.type == "Min":
                # In min heap, parent should be smaller than child
                if self.heap[index] > self.heap[leftIndex]:
                    # Swap parent with left child
                    temp = self.heap[index]
                    self.heap[index] = self.heap[leftIndex]
                    self.heap[leftIndex] = temp
                return  # No more children to check
            else:  # Max heap
                # In max heap, parent should be larger than child
                if self.heap[index] < self.heap[leftIndex]:
                    # Swap parent with left child
                    temp = self.heap[index]
                    self.heap[index] = self.heap[leftIndex]
                    self.heap[leftIndex] = temp
                return  # No more children to check
        
        # Case 3: Node has both left and right children
        else:
            if self.type == "Min":
                # Find the smaller child to potentially swap with
                if self.heap[leftIndex] < self.heap[rightIndex]:
                    swapChild = leftIndex   # Left child is smaller
                else:
                    swapChild = rightIndex  # Right child is smaller
                
                # Check if parent violates min heap property
                if self.heap[index] > self.heap[swapChild]:
                    # Swap parent with the smaller child
                    temp = self.heap[index]
                    self.heap[index] = self.heap[swapChild]
                    self.heap[swapChild] = temp
            else:  # Max heap
                # Find the larger child to potentially swap with
                if self.heap[leftIndex] > self.heap[rightIndex]:
                    swapChild = leftIndex   # Left child is larger
                else:
                    swapChild = rightIndex  # Right child is larger
                
                # Check if parent violates max heap property
                if self.heap[index] < self.heap[swapChild]:
                    # Swap parent with the larger child
                    temp = self.heap[index]
                    self.heap[index] = self.heap[swapChild]
                    self.heap[swapChild] = temp
        
        # Recursively heapify the affected subtree
        # Continue the process down the tree from where we swapped
        self.heapifyTreeExtract(swapChild)
    
    def extractNode(self):
        """
        Remove and return the root element (min/max depending on heap type).
        This operation maintains the heap property by:
        1. Replacing root with last element
        2. Removing the last element
        3. Heapifying down from root to restore heap property
        """
        # Check if heap is empty
        if self.size == 0:
            return "Heap is empty"
        
        # Step 1: Store the root value to return (this is the min/max element)
        extractedNode = self.heap[1]
        
        # Step 2: Replace root with the last element in the heap
        self.heap[1] = self.heap[self.size]
        
        # Step 3: Remove the last element (set to None for clarity)
        self.heap[self.size] = None
        
        # Step 4: Decrease the heap size
        self.size -= 1
        
        # Step 5: Restore heap property by heapifying down from root
        # Only heapify if there are still elements in the heap
        if self.size > 0:
            self.heapifyTreeExtract(1)
        
        # Step 6: Return the extracted root value
        return extractedNode
    
    def deleteHeap(self):
        """
        Delete the entire heap by resetting all properties.
        """
        self.heap = None  # Remove reference to heap array
        self.size = 0     # Reset size to 0
        return "Heap deleted successfully"


# Example usage with detailed comments
heap = Heap(5, heapType="Max")  # Create a max heap with capacity 5

# Insert elements - each will trigger heapify up to maintain max heap property
heap.insertNode(10)  # Root: 10
heap.insertNode(5)   # Tree: 10 (root), 5 (left child)
heap.insertNode(20)  # Tree: 20 (root), 5, 10 (after heapify up)

print("Heap after insertions:")
heap.levelOrderTraversal()  # Should print: 20 5 10

# Extract root (maximum element) - triggers heapify down
extracted = heap.extractNode()  # Returns 20, heapifies remaining elements
print(f"Extracted: {extracted}")

print("Heap after extraction:")
heap.levelOrderTraversal()  # Should print the remaining elements in heap order
````

## Trie
A **Trie** (prefix tree) is a tree-like data structure used to efficiently store and retrieve keys in a dataset of strings. Each node represents a character, and paths from the root to leaves represent words.
### Time and Space Complexity of Trie Operations
| Operation      | Time Complexity | Space Complexity |
|----------------|-----------------|------------------|
| Insertion      | O(m)           | O(m)             |
| Search         | O(m)           | O(1)             |
| Deletion       | O(m)           | O(1)             |
- *m is the length of the word being inserted, searched, or deleted.*
### Example of Trie
```
          (root)
         /   |   \
        c    b    a
       / \    \
      a   o    a
     /     \    \
    t       g    t
```

### Deletion from Trie
4 cases to consider when deleting a word from a Trie:
1. **Word not found**: If the word does not exist in the Trie, do nothing.
2. **Word is a prefix of another word**: If the word to be deleted is a prefix of another word, simply unmark the end-of-word flag for that node.
3. **Word has no children**: If the word to be deleted has no children, remove the node and recursively check if the parent node can also be deleted.
4. **Word has children**: If the word to be deleted has children, simply unmark the end-of-word flag for that node without deleting any nodes.
```

````python
class TrieNode:
    def __init__(self):
        self.children = {}
        self.endOfString = False
    
class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self,word):
        current = self.root
        for i in word:
            ch = i
            node = current.children.get(ch)
            if node == None:
                node = TrieNode()
                current.children[ch] = node
            current = node
        current.endOfString = True
        print(f"{word} inserted successfully")
    
    def search(self,word):
        current = self.root
        for i in word:
            node = current.children.get(i)
            if node == None:
                return f"{word} not found"
            current = node
        if current.endOfString:
            return True
        else:
            return f"{word} not found"
    
    def delete(self, word):
        """
        Public delete method that calls the recursive helper.
        """
        # Call the helper, starting from the root, at index 0
        self._delete(self.root, word, 0)
        print(f"Attempted deletion of {word}")

    def _delete(self, current_node, word, index):
        """
        Recursive helper for deletion.
        Returns True if the current_node can be deleted from its parent.
        """
        
        # --- Base Case: We have reached the end of the word ---
        if index == len(word):
            # We are at the final node for the word.
            # If it's not marked as an end, the word didn't exist anyway.
            if not current_node.endOfString:
                return False # Word not found
            
            # This is Case 1 (Word is a prefix, like "app" in "appl")
            # Un-mark it as the end of a word.
            current_node.endOfString = False
            
            # Can this node be physically deleted?
            # Only if it has no children (i.e., it's not a prefix).
            # If len(children) is 0, this returns True.
            # If len(children) > 0 (like 'l' in "appl"), this returns False.
            return len(current_node.children) == 0

        # --- Recursive Step: Move down the trie ---
        
        # Get the character and the child node
        ch = word[index]
        child_node = current_node.children.get(ch)

        # If the path doesn't exist, the word isn't here.
        if child_node == None:
            return False # Word not found

        # Recurse down. Ask the child: "Can you be deleted?"
        should_delete_child = self._delete(child_node, word, index + 1)

        # --- Post-Recursion (Backtracking): Decide what to do ---
        
        if should_delete_child:
            # The child node said it's safe to delete.
            # So, we remove it from the current node's children.
            current_node.children.pop(ch)
            
            # Now, can *this* current_node also be deleted?
            # Yes, IF:
            # 1. It has no *other* children.
            # 2. It is *not* the end of another word (e.g., "a" in "app").
            return len(current_node.children) == 0 and not current_node.endOfString
        
        # If the child should not be deleted, then this node
        # *definitely* should not be deleted.
        return False


newTrie = Trie()
newTrie.insert("app")
newTrie.insert("appl")
newTrie.delete("app")
print(newTrie.search("app"))
````
